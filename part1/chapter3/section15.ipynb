### SECTION15. 컬럼삭제, 행단위 합계, 필터링

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.주어진 데이터에서 문자 자료형 컬럼을 삭제하시오.

```python
cols = df.select_dtypes(exclude='object').columns #include(해당 타입만 탐색), exclude(해당 타입 제외)
df = df[cols]
df.head()
```

2.숫자 자료형 컬럼의 결측치를 0으로 대체하시오.

```python
df = df.fillna(0)
df.head()
```

3.각 행의 합이 3000보다 큰 값의 개수를 정수로 구하시오.

```python
df['sum'] = df.sum(axis=1, numeric_only=True)
df.head()

cond = df['sum'] > 3000
print("조건에 맞는 개수 :", sum(cond))
```
