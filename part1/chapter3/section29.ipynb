### SECTION29. 함수, 월별집계

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.배달료 계산 기준표에 따라 각 주문에 대한 배달료를 산정하시오.
##### 5km 미만 : 2000
##### 5이상 10미만 : 4000
##### 10이상 15미만 : 6000 
##### 15이상 20미만 : 8000 

```python
def delivery_fee(distince):
    fee = 0
    
    if distince < 5:
      fee = 2000
    elif distince < 10:
      fee = 4000
    elif distince < 15:
      fee = 6000
    elif distince < 20:
      fee = 8000
        
    return fee
```

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])

df['배달료'] = df['거리'].apply(delivery_fee)
df.head()
```

2.연도-월별로 배달료의 총합을 집계하시오.

```python
df['주문연월'] = df['주문시간'].dt.to_period('M')

sum_df = df.groupby(['주문연월'])['배달료'].sum()
sum_df
```

3.배달료가 가장 많이 발생한 연월을 찾고, 그 월의 총 배달료를 정수로 구하시오.

```python
max_ym = sum_df.idxmax()
max_fee = sum_df.max()

print("가장 많이 발생한 연월 :", max_ym)
print("배달료 :", max_fee)
```
