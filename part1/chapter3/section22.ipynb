### SECTION22. 시간 간의 차이 계산(분), 그룹핑

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.주문시간과 실제 도착 시간의 차이를 분 단위로 계산하시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])

df['diff'] = (df['실제도착시간']-df['주문시간']).dt.total_seconds()/60

df.head()
```

2.앱 종류별로 평균 도착시간(분)을 계산하시오.

```python
delivery_df = df.groupby(['앱종류'])['diff'].mean()
delivery_df
```

3.평균적으로 가장 빠른 앱 종류를 찾고, 해당 앱의 평균 도착 시간을 분으로 반올림하여 정수로 구하시오.

```python
delivey_df = delivery_df.sort_values()
min_diff = round(delivery_df.iloc[0])

print("가장 빠른 앱 :", delivery_df.index[0])
print("평균 도착 시간(분) :", min_diff)
```
