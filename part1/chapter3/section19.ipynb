### SECTION19. 결측치(뒤의 값으로 대체), 그룹합

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.결측치를 뒤의 값으로 대체하시오.

```python
df = df.bfill()
```

2.'city'와 'f2'를 기준으로 그룹합을 계산하시오. 

```python
sum_df = df.groupby(['city', 'f2']).sum(numeric_only=True).reset_index()
sum_df.head()
```

### 3.'views'값이 세번째로 큰 city 이름을 구하시오.

```python
sum_df = sum_df.sort_values('views', ascending=False)
city = sum_df.iloc[2,0]
print("3번째로 큰 도시 :", city)
```
