### SECTION30. 주말, 평일 구분

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.주말 주문 건수와 평일 주문 건수를 구하시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['요일']  = df['주문시간'].dt.dayofweek
df.head()

cond1 = df['요일'] < 5
cond2 = df['요일'] >= 5

print("평일 건수 :", sum(cond1))
print("주말 건수 :", sum(cond2))
```

2.주말 주문 건수와 평일 주문 건수의 차이를 절대값으로 구하고 정수로 구하시오.

```python
result = abs(sum(cond1) - sum(cond2))
print("차이값 :", result)
```
