### SECTION09. 필터링, 카테고리, 최빈값

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'views'컬럼 값이 1000이하인 데이터(결측치 제외)를 찾으시오.

```python
cond = df['views'] <= 1000
df = df[cond]
```

2.앞에서 구한 데이터 중 'f4' 컬럼의 최빈값을 구하시오.

```python
mode_val = df['f4'].mode()[0]
print("최빈값 :", mode_val)
```
