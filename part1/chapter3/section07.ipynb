### SECTION07. 값 변경(연산), 필터링 절대값

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.모든 나이(age)에 1을 더하시오.

```python
df['age'] = df['age']+1
```

2.20대의 'views'평균과 30대의 'views'평균의 차이의 절댓값을 구하시오.(반올림 후 소수점 둘째 자리까지 계산)

```python
cond = df['age'].between(20,29)
mean20 = df[cond]['views'].mean()

cond = df['age'].between(30,39)
mean30 = df[cond]['views'].mean()

diff = abs(mean20 - mean30)
diff = round(diff,2)
print("차이의 절대값 :", diff)
```
