#### SECTION14. 중복 데이터 제거, 값 변경, 데이터 개수

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.중복 데이터 제거하시오.

```python
df = df.drop_duplicates()
```

2.'f3'컬럼의 결측치는 0, 'silver'는 1, 'gold'는 2, 'vip'는 3으로 변환하시오.

```python
mapping = {
    np.nan : 0,
    'silver' : 1,
    'gold' : 2,
    'vip' : 3,
}

df['f3'] = df['f3'].map(mapping)
df.head()
```

3.변환된 'f3' 컬럼의 총합을 구하시오.

```python
sum_val = df['f3'].sum()
print("총합 : ", sum_val)
```
