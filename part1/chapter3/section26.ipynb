### SECTION26. 날짜와 시간 정보 변환, 비율

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.주문이 가장 많이 발생한 연-월을 찾으시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['주문연월'] = df['주문시간'].dt.to_period('M')
df.head()
```

```python
cnt_df = df['주문연월'].value_counts()
max_ym = cnt_df.idxmax()
max_cnt = cnt_df.max()

print("주문이 가장 많은 연월 :", max_ym)
print("주문수 :", max_cnt)
```

2.해당 연-월에 '배고팡'앱을 통한 주문 중 '앱결제'로 결재된 주문의 비율을 계산하시오.(반올림 후 소수 둘째 자리까지 계산)

```python
cond1 = df['주문연월'] == max_ym
cond2 = df['앱종류'] == '배고팡'

df = df[cond1 & cond2]

df.head()
```

```python
df['앱결제여부'] = df['결제종류'] == '앱결제'
mean_val = df['앱결제여부'].mean()

print("주문비율 :", round(mean_val,2))
```
