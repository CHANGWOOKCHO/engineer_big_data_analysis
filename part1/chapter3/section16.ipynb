### SECTION16. 이상치, IQR

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'views'컬럼의 1사분위수, 3사분위수 그리고 IQR을 계산하시오.

```python
quantile1 = df['views'].quantile(0.25)
quantile3 = df['views'].quantile(0.75)
iqr = quantile3 - quantile1

print("1사분위 :", quantile1)
print("3사분위 :", quantile3)
print("IQR :", iqr)
```

2.이상치 조건에 맞는 데이터를 찾으시오.(1사분위수-(IQR x 1.5), 3사분위수 - (IQR x 1.5))

```python
cond1 = df['views'] < quantile1-(iqr*1.5)
cond2 = df['views'] > quantile3+(iqr*1.5)

df = df[cond1 | cond2]
```

3.이상치 데이터의 'views'컬럼의 합을 정수로 구하시오.

```python
sum_val = int(df['views'].sum())
print("정수 합 :", sum_val)
```
