### 값 변경, 정렬, 합계

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

'views'컬럼의 결측 데이터를 0으로 대체하시오.

```python
df['views'] = df['views'].fillna(0)
```

'views' 컬럼에서 상위 10번째 값을 구하시오.

```python
df = df.sort_values('views', ascending = False)
df.head(10)
```

'views'컬럼에서 상위 10개의 값을 상위 10번째 값으로 대체하시오.

```python
target_val = df.iloc[9,-1]
df.iloc[0:10,-1] = target_val
```

'views'컬럼의 전체 합을 정수로 구하시오.

```python
sum_val = df['views'].sum()
print("합계 :", int(sum_val))
```
