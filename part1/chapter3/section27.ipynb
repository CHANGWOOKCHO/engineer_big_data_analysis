### SECTION27. 시간범위,속도

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.점심시간(10시 ~13시 전까지) 주문된 배달 데이터를 찾으시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])

df['시간'] = df['주문시간'].dt.hour

df.head()
```

```python
cond = df['시간'].between(10,12)
df = df[cond]
```

2.점심시간 주문 건 중 과속(평균속도 50km/h 이상)하는 주문 수를 정수로 구하시오.

#### 배달시간 = 실제도착시간 - 주문시간
#### 속도(km/h) = 거리(km)/시간(h)

```python
df['배달시간'] = (df['실제도착시간'] - df['주문시간']).dt.total_seconds()/60/60
df['속도'] = df['거리']/df['배달시간']

df.head()
```

```python
cond = df['속도'] >= 50
print("과속건수 :", sum(cond))
```
