### SECTION08. 시계열 데이터, 필터링, 데이터 개수

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'subscribed'컬럼이 2024년 2월인 데이터를 찾으시오.

```python
df['subscribed'] = pd.to_datetime(df['subscribed'], format='%Y-%m-%d')
df['new'] = df['subscribed'].dt.strftime('%Y-%m')

cond = df['new'] == '2024-02'
df = df[cond]
```

2.위에서 찾은 데이터 중 'f3'컬럼이 gold인 데이터의 개수를 구하시오.

```python
cond = df['f3'] == 'gold'
print("대상 갯수 :", sum(cond))
```
