### SECTION05. 문자열 슬라이싱, 파생변수, 평균값

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'f4' 컬럼 데이터에 'FJ'가 포함된 데이터를 찾으시오. 

```python
cond = df['f4'].str.contains('FJ')
df = df[cond]
```

2.찾은 데이터 중에서 'f2'컬럼의 평균값을 구하시오.(반올림 후 소수 둘째 자리까지 계산)

```python
mean_val = df['f2'].mean()
print("평균값 :", round(mean_val,2))
```
