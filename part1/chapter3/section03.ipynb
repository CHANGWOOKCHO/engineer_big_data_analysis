### SECTION03.파생변수, 정렬, 인덱스

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.결측치가 있는 데이터(행)을 제거하시오.

```python
df = df.dropna()
```

2.'views'컬럼을 'f1'컬럼으로 나눈 값을 새로운 컬럼으로 추가하시오.

```python
df['new'] = df['views']/df['f1']
```

3.새로운 컬럼 값 중 가장 큰 값을 가진 행의 age를 정수로 구하시오.

```python
max_val = df['new'].max()
cond = df['new'] == max_val
df = df[cond]
df
```

```python
age = int(df.iloc[0,1])
print("age :", age)
```
