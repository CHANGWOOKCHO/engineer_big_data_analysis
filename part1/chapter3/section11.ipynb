### SECTION11. 슬라이싱, 사분위수, 결측치 제거

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.데이터에서 결측치가 있는 데이터(행)을 모두 제거하시오.

```python
df = df.dropna()
```

2.결측치가 제거된 데이터를 사용하여 앞에서부터 70%데이터를 구하시오.(단, 데이터 70%지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)

```python
index = int(len(df) * 0.7)
df = df.iloc[0:index]
```

3.앞에서 구한 70% 데이터 중 'views'컬럼의 3사분위수에서 1사분위수를 뺀 값을 소수점 이하를 버리고 정수 부분만 구하시오.

```python
quantile3 = df['views'].quantile(0.75)
quantile1 = df['views'].quantile(0.25)

iqr = int(quantile3 - quantile1)
print('IQR :', iqr)
```
