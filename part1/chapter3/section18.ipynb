### SECTION18. 데이터(행) 기준 평균값, 인덱싱

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data2.csv', index_col='year')
df.head()
```

1.index '2001' 데이터(행)에서 평균보다 큰 값의 개수를 구하시오.

```python
m1 = df.loc[2001].mean()
cond1 = df.loc[2001] > m1
r1 = sum(cond1)
print("2001년 평균보다 큰 값의 개수 :", r1)
```

2.index '2003' 데이터(행)에서 평균보다 작은 값의 개수를 구하시오.

```python
m2 = df.loc[2003].mean()
cond2 = df.loc[2003] < m2
r2 = sum(cond2)
print("2003년 평균보다 작은 값의 개수 :", r2)
```

3.두 개수를 더하시오.

```python
print("result :", r1+r2)
```
