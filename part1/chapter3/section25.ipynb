### SECTION25. 시간 간의 차이 계산(일)

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.각 사용자별 첫 주문과 마지막 주문 사이의 시간 간격을 일단위로 계산하시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])

first_df = df.groupby(['user'])['주문시간'].min()
last_df = df.groupby(['user'])['주문시간'].max()

first_df.head()
```

```python
last_df.head()
```

```python
order_df = pd.DataFrame({
    '최초주문' : first_df,
    '마지막주문' : last_df,
})
order_df['diff'] = (order_df['마지막주문'] - order_df['최초주문']).dt.total_seconds()/60/60/24
order_df.head()
```

2.시간 차이가 0일인 사용자를 제외하고, 나머지 사용들의 평균 시간 간격(일 단위를 계산하시오.)

```python
cond = order_df['diff'] > 0
order_df = order_df[cond]

mean_val = order_df['diff'].mean()

print("평균 주문간격 :", mean_val)
```

3.평균 시간 간격보다 긴 시간 간격을 가진 사용자의 수를 정수로 구하시오.

```python
cond = order_df['diff'] > mean_val
print("평균보다 긴 주문 간격을 가진 사용자의 수 :", sum(cond))
```
