### SECTION12. 결측치 처리, 최빈값, 데이터 개수

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.결측치가 가장 많은 두 컬럼을 찾으시오.

```python
df.isnull().sum()
```

2.첫 번째로 결측치가 많은 컬럼에서 결측치가 있는 데이터(행)을 삭제하시오.

```python
df = df.dropna(subset=['f1'])
```

3.두 번째로 결측치가 많은 컬럼을 최빈값으로 대체하시오.

```python
mode = df['f3'].mode()[0]
df['f3'] = df['f3'].fillna(mode)
```

4.'f3'컬럼의 'gold'값을 가진 데이터의 수를 정수형으로 구하시오.

```python
cond = df['f3'] == 'gold'
print("데이터의 수 :", sum(cond))
```
