### SECTION24. 그룹핑, 값찾기, 필터링

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.사용자별 주문 거리 합계가 50km 이상인 사람들의 결제 방식을 구하시오.

```python
sum_df = df.groupby(['user'])['거리'].sum()
cond = sum_df >= 50
sum_df = sum_df[cond]
sum_df.head()
```

2.이 결제 방식 중 가장 빈도가 높은 수를 구하시오.

```python
cond = df['user'].isin(sum_df.index)
df = df[cond]
df.head()
```

```python
result = df['결제종류'].value_counts()

print("빈도가 가장 큰 결제종류 :", result.idxmax())
print("빈도가 가장 큰 수 :", result.max())
```
