### SECTION13. 결측 데이터 찾기, 필터링, 평균값

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'f1'컬럼에 결측치가 있는 데이터만 선택하시오.

```python
cond = df['f1'].isnull()
df =df[cond]
```

2.선택된 데이터에서 'age'컬럼의 평균값을 구하시오.(반올림 후 첫째 자리까지 계산) 

```python
mean_val = round(df['age'].mean(),1)
print("age 평균값 :",mean_val)
```
