### SECTION20.시계열 데이터, 월별집계, 인덱스

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.연도 구분 없이 월별로 숫자형 컬럼의 합을 구하시오.

```python
df['subscribed'] = pd.to_datetime(df['subscribed'], format='%Y-%m-%d')
df['month'] = df['subscribed'].dt.month
df.head()
```

```python
sum_df = df.groupby(['month']).sum(numeric_only=True)
sum_df.head()
```

2.합계 중 'views'가 가장 작은 값을 가진 월을 정수로 구하시오.

```python
sum_df = sum_df.sort_values('views')
print("views가 가장 작은 월 :", sum_df.index[0])
```
