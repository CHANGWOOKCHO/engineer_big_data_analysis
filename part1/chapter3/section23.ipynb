### SECTION23. 시간 간의 차이 계산(분), 비율

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.각 결제 종류별로 실제 도착시간이 예상 도착 시간보다 늦은 주문의 비율을 계산하시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'])

df['diff'] = (df['실제도착시간'] - df['예상도착시간']).dt.total_seconds()/60
df['지연여부'] = df['diff'] > 0

mean_df = df.groupby(['결제종류'])['지연여부'].mean()

mean_df
```

2.비율 중 가장 큰 값을 반올림하여 소수 둘째 자리까지 구하시오.

```python
maxidx = mean_df.idxmax()
max_val = mean_df.max()

print("비율이 가장 큰 결제방식 :", maxidx)
print("비율이 가장 큰 값 :", round(max_val,2))
```
