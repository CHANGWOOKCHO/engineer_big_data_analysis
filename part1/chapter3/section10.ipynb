### SECTION10. 그룹핑, 최댓값, 정렬

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.결측치 행을 삭제하시오.

```python
df = df.dropna()
```

2.결측치가 삭제된 데이터를 사용하여 지역별(city)평균을 계산하시오.

```python
df = df.groupby(['city']).mean(numeric_only=True)
df
```

3.앞에서 계산한 지역별 평균 데이터에서 'f2'컬럼 값이 가장 큰 지역을 구하시오.

```python
df = df.sort_values('f2', ascending=False)
print("f2가 가장 큰 지역 :",df.index[0])
```

```python
city = df['f2'].idxmax()
print("f2가 가장 큰 지역 :", city)
```
