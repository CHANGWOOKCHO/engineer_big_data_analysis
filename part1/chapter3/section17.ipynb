### SECTION17.이상치, 소수점 있는 데이터 찾기, 표준편차

```python
import pandas as pd
```

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

1.'views'컬럼의 표준편차를 구하시오.

```python
std = df['views'].std()
print("표준편차 :", std)
```

2.'age'컬럼의 이상치(소수점 나이, 음수 나이, 0살)을 제거하고, 'views'컬럼의 표준편차를 구하시오.

```python
cond = df['age'] > 0
df = df[cond]

cond = df['age']%1 == 0
df = df[cond]
```
