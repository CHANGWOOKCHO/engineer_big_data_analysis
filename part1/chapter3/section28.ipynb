### SECTION28. 날짜와 시간, 문자열

```python
import pandas as pd
```

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

1.연도와 월을 기준으로 주문 수를 집계하시오.

```python
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['주문연월'] = df['주문시간'].dt.to_period('M')

cnt_df = df['주문연월'].value_counts()
cnt_df
```

```python
max_ym = cnt_df.idxmax()
max_cnt = cnt_df.max()

print("주문이 가장 많은 연월 :", max_ym)
print("주문 수 :", max_cnt)
```

2.가장 많이 주문이 있었던 연도와 월을 예시와 같은 형식으로 숫자로만 구하시오.(2024년 2월 -> 202402)

```python
str_ym = str(max_ym)
str_ym = str_ym.replace('-','')
print("연월 :", str_ym)
```
