## S22.시간 간의 차이 계산(분), 그룹핑

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.주문 시간과 실제 도착 시간의 차이를 분 단위로 계산하시오.

```python
df["배달시간"] = df['실제도착시간'] - df['주문시간']
df["배달시간"] = df['배달시간'].dt.total_seconds()/60
df.head()
```

#### 2.앱 종류별로 평균 도착 시간(분)을 계산하시오.

```python
mean_df = df.groupby('앱종류').mean(numeric_only=True)
mean_df.head()
```

#### 평균적으로 가장 빠른 앱 종류를 찾고, 해당 앱의 평균 도착 시간을 분으로 반올림하여 정수로 구하시오. 

```python
mean_df = mean_df.sort_values('배달시간', ascending=True)
mean_df = mean_df.reset_index(drop=False)

mean_df
```

```python
print('가장 빠른 앱 :', mean_df.iloc[0,0])
print('평균도착시간(분) :', round(mean_df.iloc[0,-1]))
```

## S23.시간 간의 차이 계산(분), 비율

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 각 결제 종류별로 실제 도착 시간이 예상 도착 시간보다 늦은 주문의 비율을 계산하시오.

```python
df['지연시간'] = (df['실제도착시간'] - df['예상도착시간']).dt.total_seconds()/60
df['지연여부'] = df['지연시간'] > 0

result = df.groupby(['결제종류'])["지연여부"].mean()

result
```

#### 비율 중 가장 큰 값을 반올림하여 소수 둘째 자리까지 구하시오.

```python
print("비율이 가장 큰 결제종류 :", result.idxmax())
print("비율이 가장 큰 값:", round(result.max(),2))
```

## S25.시간 간의 차이 계산(일)

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.각 사용자별로 첫 주문과 마지막 주문 사이의 시간 간격을 일 단위로 계산하시오.

```python
user_min_order = df.groupby(['user'])['주문시간'].min()
user_max_order = df.groupby(['user'])['주문시간'].max()

user_min_order.name = 'min'
user_max_order.name = 'max'

order_df = pd.concat([user_min_order, user_max_order], axis=1)
order_df['시간간격'] = (order_df['max'] - order_df['min']).dt.days 
#order_df['test'] = (order_df['max'] - order_df['min']).dt.total_seconds()/60/60/24
order_df.head()
```

#### 2.시간차가 0일인 사용자를 제외하고, 나머지 사용자들의 평균 시간 각격(일 단위)을 계산하시오.

```python
cond = order_df['시간간격'] > 0
order_df = order_df[cond]

mean_interval = order_df['시간간격'].mean()
print("mean_interval :", mean_interval)
```

#### 3.평균 시간 간격보다 긴 시간 간격을 가진 사용자의 수를 정수로 구하시오.

```python
cond = order_df['시간간격'] > mean_interval
print("사용자 수 :", sum(cond))
```

## S26.날짜와 시간 정보 변환, 비율

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.주문이 가장 많이 발생한 연-월을 찾으시오. 

```python
df['주문연월'] = df['주문시간'].dt.to_period('M')

order_max = df['주문연월'].value_counts().idxmax()
order_count= df['주문연월'].value_counts().max()

print("최대 주문발생 연월 :", order_max)
print("최대 주문발생 건수 :", order_count)
```

#### 2.해당 연-월에 '배고팡'앱을 통한 주문 중 '앱결제'로 결제된 주문의 비율을 계산하시오.
#### (반올림 후 소수 둘째 자리까지 계산)

```python
cond1 = df['주문연월'] == order_max
cond2 = df['앱종류'] == '배고팡'

filtered_df = df[cond1 & cond2]

filtered_df.head()
```

```python
cond = filtered_df['결제종류'] == '앱결제'
result = sum(cond) / len(filtered_df)

print("앱결제 비율 :", round(result,2))
```

## S27.시간 범위, 속도(km/h)

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.점심시간(10시부터 13시 전까지)에 주문된 배달 데이터를 찾으시오.

```python
df['시간'] = df['주문시간'].dt.hour
df.head()

cond = df['시간'].between(10,13, inclusive='left')
print("점심시간 주문건수 :", sum(cond))

df = df[cond]
df.head()
```

#### 2.점심시간 주문 건 중 과속(평균 속도가 50km/h 이상)하는 주문 수를 정수로 구하시오.
#### 배달시간=실제도착시간 - 주문시간
#### 속도(km/h)=거리(km)/시간(h)

```python
df['배달시간'] = df['실제도착시간'] - df['주문시간']
df['배달시간'] = df['배달시간'].dt.total_seconds()/60/60
df['속도'] = df['거리']/df['배달시간']

df.head()
```

```python
cond = df['속도'] >= 50
print('과속 건수 :', sum(cond))
```

## S29.함수, 월별 집계

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.배달료 계산 기준표에 따라 각 주문에 대한 배달료를 계산하시오.
##### 5km 미만 : 2000원
##### 5km 이상 ~ 10km 미만 : 4000원
##### 10km 이상 ~ 15km 미만 : 6000원
##### 15km 이상 ~ 20km 미만 : 8000원

```python
def delivery_fee(distance):
    result = 0
    if distance < 5:
        result = 2000
    elif distance < 10:
        result = 4000
    elif distance < 15:
        result = 6000
    elif distance < 20:
        result = 8000
    return result
```

```python
df['배달료'] = df['거리'].apply(delivery_fee)
df.head()
```

#### 2.연도-월별로 배달료의 총합을 집계하시오.

```python
df['배달연월'] = df['주문시간'].dt.to_period('M')
sum_df = df.groupby(['배달연월'])['배달료'].sum()

sum_df
```

#### 3.배달료가 가장 많이 발생한 월을 찾고, 그 월의 총 배달료를 정수로 구하시오.

```python
print("배달료가 가장 많이 발생한 월 :", sum_df.idxmax())
print("총배달료 :", sum_df.max())
```

## 주말, 평일구분

```python
df = pd.read_csv('delivery_time.csv')
df['주문시간'] = pd.to_datetime(df['주문시간'], format='%Y-%m-%d %H:%M:%S')
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'], format='%Y-%m-%d %H:%M:%S')
df['예상도착시간'] = pd.to_datetime(df['예상도착시간'], format='%Y-%m-%d %H:%M:%S')

df.head()
```

#### 1.주말주문건수와 평일 주문 건수를 구하시오.

```python
df['요일'] = df['주문시간'].dt.dayofweek
df['주말여부'] = df['요일'] >= 5

weekend = sum(df['주말여부'])
weekday = sum(~df['주말여부'])

print("평일 수  :", weekday)
print("주말 수  :", weekend)
```

#### 주말주문건수와 평일주문건수의 차이를 절대값으로 구하고 정수형으로 구하시오.

```python
result = abs(weekday-weekend)
result = int(result)

print("결과 :", result)
```
