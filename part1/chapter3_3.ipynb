## S31.문자열, 형변환

```python
df = pd.read_csv('delivery_time.csv')
df.head()
```

#### 1.'user'컬럼에서 user뒤에 있는 숫자 값만 추출하시오.

```python
df['user_num'] = df['user'].replace(r'\D','',regex=True)
df.head()
```

#### 2.추출된 숫자 값을 모두 합한 값을 정수로 구하시오.

```python
df['user_num'] = df['user_num'].astype('int')

result = int(df['user_num'].sum())
print("합 결과 :", result)
```

## S32.합계(열 방향), 상위 값 선택

```python
df = pd.read_csv('school_data.csv')
df.head()
```

#### 1.수학,영어,국어 점수의 합을 구하시오.

```python
df['total_score'] = df.sum(numeric_only=True, axis=1)
df.head()
```

#### 2.합이 가장 큰 상위 10명을 찾으시오.

```python
df = df.sort_values('total_score', ascending=False)
df = df.iloc[0:10]
```

#### 3.찾은 10명의 수학 평균 점수를 구하시오.(반올림 후 정수 출력)

```python
result = df['수학'].mean()
result = round(result)

print('수학 평균 :', result)
```

## S33. 데이터프레임 재구조화

```python
df = pd.read_csv('school_data.csv')
df.head()
```

#### 1.과목에 상관없이 점수가 가장 작은 점수 25개를 찾으시오.

```python
df = df.melt(id_vars=['이름'], value_vars=['수학','영어','국어'], var_name='과목', value_name='점수')
df = df.nsmallest(25, '점수')

print("데이터 수 :", len(df))
df.head()
```

## S35.데이터 합치기(merge)

```python
df = pd.read_csv('school_data.csv')
df.head()
```

```python
social_df = pd.read_csv('school_data_social.csv')
social_df.head()
```

#### 1.두 csv파일을 '이름'을 기준으로 합치시오.

```python
df = pd.merge(df, social_df, on='이름')
df.head()
```

#### 2.영어교사가 장선생이면서 사회교사가 오선생인 학생들을 필터링하시오.

```python
cond1 = df['영어교사'] == '장선생'
cond2 = df['사회교사'] == '오선생'
df = df[cond1 & cond2]
```

#### 3.필터링된 학생들의 수학 점수를 모두 더한 후 정수를 구하시오.

```python
result = int(df['수학'].sum())
print("수학점수 합계 :", result)
```

## S36.조건별 변환

```python
df = pd.read_csv('sales.csv')
df.head()
```

#### 1.결측된 판매금액은 해당 지역의 평균 판매금액으로 결측값을 대체하시오.

```python
mean_value = df.groupby('지역코드')['판매금액'].transform('mean')
df['판매금액'] = df['판매금액'].fillna(mean_value)
df.head()
```

#### 2.각 거래마다 '판매금액'과 해당 지역의 평균 판매금액의 '차이'를 절대값으로 구하시오.

```python
df['지역평균'] = df.groupby('지역코드')['판매금액'].transform('mean')
df['평균차이'] = abs(df['판매금액'] - df['지역평균'])
df.head()
```

#### 3.각 지역에서 차이값의 평균을 구한 후, 이 값이 가장 큰 지역의 지역코드를 구하시오.

```python
result = df.groupby('지역코드')['평균차이'].mean()
print("차이가 가장 큰 지역 :", result.idxmax())
print("차이가 가장 큰 값 :", result.max())
```

## S37.재구조화(unstack), 맵핑 

```python
df = pd.read_csv('store_sales.csv')
df.head()
```

#### 1.각 행에서 판매수량과 단가를 이용하여 매출액을 계산하시오.

```python
df['매출액'] = df['판매수량'] * df['단가']
df.head()
```

#### 2.요일을 평일과 주말로 구분하고, 매장별 평일과 주말 매출액 합계를 구하시오.

```python
mapping = {
    '월':'평일',
    '화':'평일',
    '수':'평일',
    '목':'평일',
    '금':'평일',
    '토':'주말',
    '일':'주말',
    }
df['구분'] = df['요일'].map(mapping)
df.head()
```

```python
sales_df = df.groupby(['매장코드','구분'])['매출액'].sum()
sales_df
```

```python
sales_df = sales_df.unstack(level=-1)
```

```python
sales_df.head()
```

#### 3.매장별 평일과 주말 매출액 차이를 절대값으로 구하시오.
####   이후, 모든 매장 중 가장 큰 절대값 차이를 찾으시오.

```python
sales_df['매출차이'] = abs(sales_df['평일']-sales_df['주말'])
sales_df.head()
```

```python
result = sales_df['매출차이'].max()
print("가장 큰 절대값 :", result)
```

## S38.피벗테이블

```python
df = pd.read_csv('region_sales.csv')
df.head()
```

#### 1.각 Region과 Channel 조합별로 제품(Product) A,B의 총 판매액을 계산하시오.

```python
test = df.groupby(['Region', 'Channel', 'Product'])['Sales'].sum()
test.unstack() 
```

```python
pivot_table = pd.pivot_table(df, 
                             index=['Region', 'Channel'],
                             columns='Product',
                             values='Sales',
                             aggfunc='sum',
                            )
pivot_table
```

#### 제품 A의 매출 비율(A비율)을 구하시오.
#### A비율 = (제품A의 판매액) / (제품A 판매액 + 제품B 판매액)

```python
pivot_table['총판매액'] = pivot_table['A'] + pivot_table['B']
pivot_table['A비율'] = pivot_table['A']/pivot_table['총판매액']
pivot_table
```

#### 3.A비율 중 최댓값을 찾아 반올림하여 소수 둘째 자리까지 구하시오.

```python
result = pivot_table['A비율'].max()
result = round(result,2)
print("A비율 최댓값 :", result)
```
