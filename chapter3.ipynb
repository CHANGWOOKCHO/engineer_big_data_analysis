## S6.필터링, 분산

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

## S7.값 변경(연산), 필터링 절댓값

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

#### 1.모든 나이(age)에 1을 더하시오.

```python
df['age'] =df['age']+1
df.head()
```

#### 2.20대의 'views' 평균과 30대의 'views' 평균의 차이의 절대값을 구하시오.(반올림 후 소수 둘째 자리까지 계산)

```python
cond1 = df['age'].between(20,29)
cond2 = df['age'].between(30,39)

mean20 = df[cond1]['views'].mean()
mean30 = df[cond2]['views'].mean()

abs_diff = abs(mean20-mean30)
abs_diff = round(abs_diff,2)
print(abs_diff)
```

## S8.시계열데이터, 필터링, 데이터 개수

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

#### 1.'subscribed' 컬럼이 2024년 2월인 데이터를 찾으시오.

```python
df['subscribed'] = pd.to_datetime(df['subscribed'], format='%Y-%m-%d')
df.head()
```

#### 2.위에서 찾은 데이터 중 'f3' 컬럼이 gold인 데이터의 개수를 구하시오.

```python
cond1 = df['subscribed'].dt.strftime('%Y-%m') == '2024-02'
cond2 = df['f3'] == 'gold'

target_cnt = len(df[cond1&cond2])
print("대상 개수 :", target_cnt)
```

## S10.그룹핑, 최댓값, 정렬

```python
df = pd.read_csv('type1_data1.csv')
df.head()
```

#### 1.결측치가 있는 행을 삭제하시오.

```python
df = df.dropna()
df.head()
```

#### 2.결측치가 삭제된 데이터를 사용하여 지역별(city) 평균을 구하시오.

```python
mean_df = df.groupby(['city']).mean(numeric_only=True)
mean_df
```

#### 3.앞에서 계산한 지역별 평균 데이터에서 'f2'컬럼 값이 가장 큰 지역을 구하시오.

```python
mean_df = mean_df.sort_values('city', ascending=False)
print('f2 컬럼이 가장 큰 지역 :', mean_df.index[0])
print('f2 컬럼이 가장 큰 지역 :', mean_df['f2'].idxmax())
```
