```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
```

```python
data = {
    '키': [150, 160, 170, 175, 165, 155, 172, 168, 174, 158,
          162, 173, 156, 159, 167, 163, 171, 169, 176, 161],
    '몸무게': [42, 50, 70, 64, 56, 48, 68, 60, 65, 52,
            54, 67, 49, 51, 58, 55, 69, 61, 66, 53]
}
df = pd.DataFrame(data)
df.head()
```

```python
model = ols('키~몸무게',data=df).fit()
print(model.summary())
```

```python
print("결정계수 :", model.rsquared)
print("모델의 검정통계량 :", model.fvalue)
print("모델의 pvalue :", model.f_pvalue)

print("절편 :", model.params.iloc[0])
print("몸무게의 기울기 :", model.params.iloc[1])
print("몸무게의 pvalue :", model.pvalues.iloc[1])
```

```python
df['예측값'] = model.predict(df)
df['잔차'] = df['키'] - df['예측값']
df.head()
```

```python
print("회귀제곱합 :", model.ess)
print("잔차제곱합(오류제곱합) :", model.ssr)
print("총제곱합 :", model.centered_tss)
```

### 신뢰구간과 예측구간

```python
result = model.conf_int(alpha=0.5)
print(f'몸무게의 95% 신뢰구간 : {result.loc["몸무게",0]} ~ {result.loc["몸무게",1]}')
```

```python
new_data = pd.DataFrame({
    '몸무게' : [50]
})
```

```python
print("몸무게가 50일 때의 신뢰구간과 예측구간============")
pred = model.get_prediction(new_data)
result = pred.summary_frame(alpha=0.05)
result
```
