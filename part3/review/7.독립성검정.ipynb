```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
df =pd.read_csv('like.csv')
df.head()
```

```python
df = pd.crosstab(df['성별'], df['운동'])
df
```

```python
result = stats.chi2_contingency(df)
result
```

```python
print("statistic :", result.statistic)
print("pvalue :", result.pvalue)
print("===기대빈도해석===")
print(result.expected_freq)
```
