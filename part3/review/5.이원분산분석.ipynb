```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
```

```python
df = pd.read_csv('tree.csv')
df.head()
```

```python
model = ols('성장률 ~ 나무*비료', data=df).fit()
print(model.summary())
```

```python
print("결정계수 :", model.rsquared)
print("수정된 결정계수 :", model.rsquared_adj)
print("검정통계량 :", model.fvalue)
print("p-value :", model.f_pvalue)
```

```python
lm = anova_lm(model)
lm
```
