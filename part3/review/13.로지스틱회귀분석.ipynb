```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import logit
```

```python
df = pd.read_csv('health_survey2.csv')
df.head()
```

```python
model = logit('disease ~ age + bmi + smoker + activity_level + lifestyle', data=df).fit()
model.summary()
```

```python
print("로그우도 :", model.llf)
print("잔차이탈도 :", model.llf * -2)
```

```python
print('age의 오즈비 :', np.exp(model.params['age']))
print('bmi 3증가 시의 오즈비 :', np.exp(model.params['bmi']*3))
print('흡연여부의 오즈비 :', np.exp(model.params['smoker']))
```

```python
print("라이프스타일의 오즈비 =============")
print("A대비 M의 오즈비 :", np.exp(model.params['lifestyle[T.M]']))
print("A대비 S의 오즈비 :", np.exp(model.params['lifestyle[T.S]']))

print("M대비 A의 오즈비 :", np.exp(-model.params['lifestyle[T.M]']))
print("S대비 A의 오즈비 :", np.exp(-model.params['lifestyle[T.S]']))

print("M대비 S의 오지비 :", np.exp(model.params['lifestyle[T.M]']-model.params['lifestyle[T.S]']))
print("S대비 M의 오지비 :", np.exp(model.params['lifestyle[T.S]']-model.params['lifestyle[T.M]']))
```
