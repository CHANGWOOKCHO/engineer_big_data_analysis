```python
import numpy as np
import pandas as pd
```

```python
data = {
    '키': [150, 160, 170, 175, 165],
    '몸무게': [42, 50, 70, 64, 56]
}
df = pd.DataFrame(data)
```

### 피어슨 상관계수

```python
df.corr()
```

### 스피어만 상관계수

```python
df.corr(method='spearman')
```

### 켄달 상관계수

```python
df.corr(method='kendall')
```
