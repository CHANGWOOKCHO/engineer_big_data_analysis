```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
```

```python
df = pd.read_csv('study.csv')
df.head()
```

```python
model = ols('score ~ study_hours + material_type',data=df).fit()
model.summary()                                                
```

```python
print("모델의 결정계수 :", model.rsquared)
print("모델의 검정통계량 :", model.fvalue)
print("모델의 pvalue :", model.f_pvalue)
```

```python
print("회귀제곱합 :", model.ess)
print("오류제곱합 :", model.ssr)
print("총제곱합 :", model.centered_tss)
```

```python
model.conf_int(alpha=0.05)
```
