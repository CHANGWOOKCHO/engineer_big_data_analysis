```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
```

```python
df = pd.read_csv('fertilizer.csv')
df.head()
```

```python
model = ols('성장~C(비료)',data=df).fit()
print(model.summary())
```

```python
print("결정계수 :", model.rsquared)
print("수정된 결정계수 :", model.rsquared_adj)
print("검정통계량 :", model.fvalue)
print("p-value :", model.f_pvalue)
```

```python
lm = anova_lm(model)
lm
```

```python
print("전체 데이터의 자유도 :", len(df)-1)
print("그룹 자유도 :", lm.iloc[0]['df'])
print("잔차 자유도 :", lm.iloc[1]['df'])
print("검정통계량 :", lm.iloc[0]['F'])
print("p-value :", lm.iloc[0]['PR(>F)'])
```

### f_oneway로 검정

```python
from scipy import stats
```

```python
df2 = pd.DataFrame({
    'A' : df[df['비료'] == 'A']['성장'].to_numpy(),
    'B' : df[df['비료'] == 'B']['성장'].to_numpy(),
    'C' : df[df['비료'] == 'C']['성장'].to_numpy(),
    'D' : df[df['비료'] == 'D']['성장'].to_numpy(),
})

df2.head()
```

```python
statistic, pvalue = stats.f_oneway(df2['A'],df2['B'],df2['C'],df2['D'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```
