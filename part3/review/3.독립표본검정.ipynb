```python
import numpy as np
import pandas as pd
from scipy import stats 
```

```python
df = pd.read_csv('class_score.csv')
df.head()
```

### 정규성 검정 : 각각의 집단을 shapiro검정

```python
statistic, pvalue = stats.shapiro(df['A반'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

```python
statistic, pvalue = stats.shapiro(df['B반'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### 등분산성 검정 : levene검정

```python
statistic, pvalue = stats.levene(df['A반'], df['B반'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### ttest_ind() : 정규성 만족, 등분산성 만족

```python
statistic, pvalue = stats.ttest_ind(df['A반'], df['B반'], equal_var=True)
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### ttest_ind(equal_var=False) : 정규성 만족, 등분산성 불만족

```python
statistic, pvalue = stats.ttest_ind(df['A반'], df['B반'], equal_var=False)
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### mannwhitneyu : 정규성 불만족

```python
statistic, pvalue = stats.mannwhitneyu(df['A반'], df['B반'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```
