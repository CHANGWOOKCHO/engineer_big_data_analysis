```python
import numpy as np
import pandas as pd

from scipy import stats
```

```python
df= pd.read_csv('popcon.csv')
df.head()
```

### 기술통계량

```python
print("분산 :", round(df['weights'].var(),3))
print("왜도 :", round(df['weights'].skew(),3))
print("첨도 :", round(df['weights'].kurt(),3))
print("표준편차 :", round(df['weights'].std(),3))
print("표준오차 :", round(df['weights'].std()/np.sqrt(len(df)),3))
```

### 정규성검정

```python
statistic, pvalue = stats.shapiro(df['weights'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### ttest_1samp : 단일표본 정규성 만족시 사용

```python
statistic, pvalue = stats.ttest_1samp(df['weights'], 120)
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g이 아니다")
```

```python
statistic, pvalue = stats.ttest_1samp(df['weights'], 120, alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g보다 작다")
```

```python
statistic, pvalue = stats.ttest_1samp(df['weights'], 120, alternative='greater')
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g보다 크다")
```

### wilcoxon : 단일표본 정규성 불만족 시 사용

```python
statistic, pvalue = stats.wilcoxon(df['weights']-120)
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g이 아니다")
```

```python
statistic, pvalue = stats.wilcoxon(df['weights']-120, alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g보다 작다")
```

```python
statistic, pvalue = stats.wilcoxon(df['weights']-120, alternative='greater')
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("팝콘의 무게는 평균 120g이다")
else:
    print("팝콘의 무게는 평균 120g보다 크다")
```
