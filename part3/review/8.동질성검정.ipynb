```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('club.csv')
df.head()
```

```python
df = pd.crosstab(df['학과'], df['동아리가입여부'])
df
```

```python
result = stats.chi2_contingency(df)
```

```python
print("검정통계량 :", result.statistic)
print("p-value :", result.pvalue)
print("===기대빈도===")
print(result.expected_freq)
```
