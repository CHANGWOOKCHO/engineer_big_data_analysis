```python
import numpy as numpy
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('education.csv')
df.head()
```

### 정규성검정 : shapiro(차이값)

```python
statistic, pvalue = stats.shapiro(df['after']-df['before'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### ttest_rel() : 정규성 만족 시 사용

```python
statistic, pvalue = stats.ttest_rel(df['before'], df['after'])
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 전과 후의 평균이 같지 않다")
```

```python
statistic, pvalue = stats.ttest_rel(df['before'], df['after'], alternative='less') ## a-b의 차이값을 기준으로 -값을 가정한다면 'less'
                                                                                   ##                         +값을 가정한다면 'greater'
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 전보다 후의 평균이 더 높다")
```

```python
statistic, pvalue = stats.ttest_rel(df['before'], df['after'], alternative='greater') ## a-b의 차이값을 기준으로 -값을 가정한다면 'less'
                                                                                      ##                         +값을 가정한다면 'greater'
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 후보다 전의 평균이 더 높다")
```

### wilcoxon: 정규성 불만족 시

```python
statistic, pvalue = stats.wilcoxon(df['before'], df['after'])
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 전과 후의 평균이 같지 않다")
```

```python
statistic, pvalue = stats.wilcoxon(df['before'], df['after'], alternative='less') ## a-b의 차이값을 기준으로 -값을 가정한다면 'less'
                                                                                  ##                         +값을 가정한다면 'greater'
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 전보다 후의 평균이 더 높다")
```

```python
statistic, pvalue = stats.wilcoxon(df['before'], df['after'], alternative='greater') ## a-b의 차이값을 기준으로 -값을 가정한다면 'less'
                                                                                     ##                         +값을 가정한다면 'greater'
print("statistic :", statistic)
print("pvalue :", pvalue)

if pvalue > 0.05:
    print("교육 전과 후의 평균이 같다")
else:
    print("교육 전보다 후의 평균이 더 높다")
```
