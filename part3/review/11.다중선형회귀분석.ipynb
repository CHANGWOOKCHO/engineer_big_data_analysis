```python
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.formula.api import ols
```

```python
data = {
    '매출액': [300, 320, 250, 360, 315, 328, 310, 335, 326, 280,
            290, 300, 315, 328, 310, 335, 300, 400, 500, 600],
    '광고비': [70, 75, 30, 80, 72, 77, 70, 82, 70, 80,
            68, 90, 72, 77, 70, 82, 40, 20, 75, 80],
    '직원수': [15, 16, 14, 20, 19, 17, 16, 19, 15, 20,
            14, 5, 16, 17, 16, 14, 30, 40, 10, 50]
    }
df = pd.DataFrame(data)
df.head()
```

```python
model = ols('매출액~광고비+직원수',data=df).fit()
model.summary()
```

```python
print("광고비의 상관계수 :", df['광고비'].corr(df['매출액']))
print("직원수의 상관계수 :", df['직원수'].corr(df['매출액']))
```

```python
stats.pearsonr(df['광고비'], df['매출액'])
```

```python
print("회귀제곱합 :", model.ess)
print("잔차제곱합 :", model.ssr)
print("총제곱합 :", model.centered_tss)
```

```python
print("절편 :", model.params.iloc[0])
print("광고비의 기울기 :", model.params.loc['광고비'])
print("직원수의 기울기 :", model.params.loc['직원수'])
```

### 변수의 신뢰구간

```python
model.conf_int(alpha=0.05)
```

### 예측값의 신뢰구간과 예측구간

```python
new_data = pd.DataFrame({
    '광고비' : [45],
    '직원수' : [22],
})
```

```python
pred = model.get_prediction(new_data)
result = pred.summary_frame(alpha=0.05)
result
```
