```python
import pandas as pd
from scipy import stats
```

어떤 도시에서 300명을 대상으로 아이스크림 맛 선호도를 조사했다. 이 도시에서 조사된 아이스크림 맛 선호도는 바닐라:150명, 초코:120명, 딸기:30명이었다.
전국적인 아이스크림 맛 선호도 조사 결과를 통해 알려진 비율은 바닐라 50%, 초코 35%, 딸기 15%로 알려져 있다. 유의수준 0.05에서 이 도시의 아이스크림 맛 선호도는 전국적인 맛 선호도와 차이가 있는가?

```python
observed = [150,120,30]
expected = [0.5*300, 0.35*300, 0.15*300]

statistic, p_value = stats.chisquare(observed, expected)

if p_value > 0.05:
    print('귀무가설 채택(전국적인 맛 선호도와 일치) :', p_value)
else:
    print('귀무가설 기각(전국적인 맛 선호도와 불일치) :', p_value)
```
