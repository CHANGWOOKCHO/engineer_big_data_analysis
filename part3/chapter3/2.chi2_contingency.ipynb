```python
import pandas as pd
from scipy import stats
```

성별에 따라 운동을 좋아하는지 조사한 결과다. 성별로 운동 선호도가 독립적인지 가설검정을 실시하시오.(유의수준 0.05)
##### 귀무가설 : 성별과 운동선호도는 독립적이다.(성별과 운동 선호도 간에는 연관성이 없다.)
##### 대립가설 : 성별과 운동선호도는 독립적이지 않다.(성별과 운동 선호도 간에는 연광성이 있다.)

```python
df = pd.DataFrame({
    'like' : [80, 90],
    'unlike' : [30, 10]},
    index=['남자', '여자'])

df.head()
```

```python
result = stats.chi2_contingency(df)

if result.pvalue > 0.05:
    print("독립적(연관성 없음) :", result.pvalue)
else:
    print("비동립적(연광성 있음) :", result.pvalue)

freq_nm = ['남자가 운동을 좋아함을 선택할 기대 수',
           '남자가 운동을 좋아하지 않음을 선택할 기대 수',
           '여자가 운동을 종아함을 선택할 기대 수',
           '여자가 운동을 좋아하지 않음을 선택할 기대 수'
          ]

cnt = 0
for i in range(len(result.expected_freq)) :
    for j in range(len(result.expected_freq[i])):
        print(freq_nm[cnt],":",result.expected_freq[i][j])
        cnt = cnt + 1
```

### 동질성 검정

```python
df = pd.DataFrame({
    '학과' : ['통계학과'] * 100 + ['컴퓨터공학과'] * 100,
    '가입여부' : ['가입'] * 50 + ['미가입'] * 50 + ['가입'] * 30 + ['미가입'] * 70,
})

df.sample(5)
```

```python
df = pd.crosstab(df['학과'], df['가입여부'])
df
```

```python
result = stats.chi2_contingency(df)

if result.pvalue > 0.05:
    print("분포, 비율이 동일함 :", result.pvalue)
else:
    print("분포, 비율이 동일하지 않음 :", result.pvalue)
```
