### SECTION05.이원 분산 분석

H토마토연구소에서는 토마토의 수확량을 늘리기 위해 세 가지 다른 비료 유형과 네 가지 다른 물 주기를 실험하고자 한다. 연구소에서는 12개의 토마토 식물을 무작위로 선택해 각 조합에 대대 반복적으로 실험을 수행했다.

```python
import pandas as pd
import numpy as np
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm
```

```python
df = pd.read_csv('tomato2.csv')
df.head()
```

```python
model = ols('수확량~C(비료유형)*C(물주기)', data=df).fit()
model.summary()
```

```python
anova_table = anova_lm(model)
anova_table
```

```python
print("비료유형에 따른 검정통계량 :", anova_table.iloc[0,3])
print("비료유형에 따른 p-value :", anova_table.iloc[0,4])
if  anova_table.iloc[0,4] > 0.05 :
    print("귀무가설 채택(비료)")
else:
    print("귀무가설 기각(비료)")
```

```python
print("물주기유형에 따른 검정통계량 :", anova_table.iloc[1,3])
print("물주기유형에 따른 p-value :", anova_table.iloc[1,4])
if  anova_table.iloc[1,4] > 0.05 :
    print("귀무가설 채택(물주기)")
else:
    print("귀무가설 기각(물주기)")
```

```python
print("비료*물주기유형에 따른 검정통계량 :", anova_table.iloc[2,3])
print("비료*물주기유형에 따른 p-value :", anova_table.iloc[2,4])
if  anova_table.iloc[2,4] > 0.05 :
    print("귀무가설 채택(비료*물주기)")
else:
    print("귀무가설 기각(비료*물주기)")
```
