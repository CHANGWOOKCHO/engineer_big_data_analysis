### SECTION06.적합도검정

한 도시의 운전자 1000명을 대상으로 교통사고 경험 수를 조사했다. 이 도시 운전자들의 교통사고 경험분포가 전국적인 경향을 따르는지 검정하시오.
##### 귀무가설 : 이 도시의 교통사고 경험 수 분포는 전국적인 경향을 따른다.
##### 대립가설 : 이 도시의 교통사고 경험 수 분포는 전국적인 경향을 따르지 않는다.

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.DataFrame(
    {
        "경험수" : [550,250,100,70,30],
        "전국적인경향" : [60,25,8,5,2],
    },    
    index =['1회','2회','3회','4회','5회'])

df.head()
```

1.이 도시의 교통사고 경험자 중 5회 이상의 비율을 0과 1사이로 구하시오.

```python
sum_val = df['경험수'].sum()

df['percent'] = df['경험수']/sum_val
df.head()
```

```python
print("비율 :",df.loc['5회', 'percent'])
```

2.이 도시 운전자들의 교통사고 경험 수 분포가 전국적인 경향을 따르는지 검정하기 위한 검정 통계량을 구하시오.

```python
df['expected'] = sum_val*(df['전국적인경향']/100)
df.head()
```

```python
statistic,pvalue = stats.chisquare(df['경험수'], df['expected'])
print("검정통계량 :", statistic)
```

3.위의 통계량에 대한 p-value를 구하시오.

```python
print("p-value :", pvalue)
```

4.위의 검정 결과를 유의수준 0.05에서 귀무가설을 기준으로 채택/기각 중 선택해 입력하시오.

```python
if pvalue > 0.05:
    print("귀무가설 채택")
else:
    print("귀무가설 기각")
```
