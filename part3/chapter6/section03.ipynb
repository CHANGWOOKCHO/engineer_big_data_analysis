### SECTION03.대응표본검정

```python
import pandas as pd
import numpy as np
from scipy import stats 
```

길벗 연구소에서는 새로 개발한 교육과정에 참여하면 학습 시간이 단축된다고 주장한다. 10명의 학생들을 대상으로 기존의 교육방법과 새로운 교육방법을 적용한 결과에 대해 유의수준 5%에서 새로운 교육 방법이 교육 시간을 단축시켰는지 검정하시오.(모집단은 정규분포를 가정한다.)
##### 귀무가설 : (새로운 교육 방법 - 기존 교육 방법) = 0
##### 대립가설 : (새로운 교육 방법 - 기존 교육 방법) < 0

```python
df = pd.DataFrame({
    'User': list(range(1, 11)),
    '기존방법': [60.4, 60.7, 60.5, 60.3, 60.8, 60.6, 60.2, 60.5, 60.7, 60.4],
    '새로운방법': [59.8, 60.2, 60.1, 59.9, 59.7, 58.4, 57.0, 60.3, 59.6, 59.8]
})

df.head()
```

1.표본평균을 구하시오.

```python
df['diff'] = df['새로운방법'] - df['기존방법']
mean_val = df['diff'].mean()

print("표본평균 :", mean_val)
```

2.위의 가설을 검정하기 위한 검정 통계량을 구하시오.

```python
#정규성 검정
statistic, pvalue = stats.shapiro(df['diff'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

```python
statistic, pvalue = stats.ttest_rel(df['새로운방법'], df['기존방법'], alternative='less')
print("statistic :", statistic)
```

3.위의 통계량에 대한 p-vale를 구하시오.

```python
print("pvalue :", pvalue)
```

4.유의수준 0.05에서 귀무가설을 기준으로 검정의 결과를 채택/기각 중 선택해 입력하시오.

```python
if pvalue > 0.05:
    print("귀무가설 채택")
else:
    print("귀무가설 기각")
```

### 정규성 만족 X : stats.wilcoxon(A데이터, B데이터, 대립가설 방향)

```python
statistic, pvalue = stats.wilcoxon(df['새로운방법'],df['기존방법'],alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)
```
