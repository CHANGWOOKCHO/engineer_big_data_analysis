### SECTION02.독립표본검정 

```python
import pandas as pd
import numpy as np
from scipy import stats
```

스마트폰 배터리의 충전 시간을 줄이기 위한 새로운 충전기를 개발했다. 개발된 충전기의 효과를 검정하기 위해 스마트폰 사용자 집단을 두 그룹으로 나눠 한 그룹에는 새로운 충전기를 제공하고, 다른 그룹에는 기존 충전기를 제공했다. 그 후 두 그룹의 평균 충전 완료 시간을 비교했다. 두 그룹 모두 충전 시간은 정규분포를 따르며 분산은 같다고 가정한다. 새로운 충전기가 더 빠르다고 할 수 있는지를 유의수준 5%하에서 검정하시오.
##### 귀무가설 : 새로운 충전 시간 >= 기존 충전 시간
##### 대립가설 : 새로운 충전 시간 < 기존 충전 시간

```python
df = pd.DataFrame({
    '충전기': ['New'] * 10 + ['Old'] * 10,
    '충전시간': [
        1.5, 1.6, 1.4, 1.7, 1.5, 1.6, 1.7, 1.4, 1.6, 1.5,
        1.7, 1.8, 1.7, 1.9, 1.8, 1.7, 1.8, 1.9, 1.7, 1.6
    ]
})

df.sample(5)
```

1.위의 가설을 검정하기 위한 검정 통계량을 구하시오.

```python
cond1 = df['충전기'] == 'New'
cond2 = df['충전기'] == 'Old'

new_df = df[cond1]
old_df = df[cond2]
```

```python
#정규성 검정
statistic, pvalue = stats.shapiro(new_df['충전시간'])
print("new그룹 statistic :", statistic)
print("new그룹 pvalue :", pvalue)

statistic, pvalue = stats.shapiro(old_df['충전시간'])
print("old그룹 statistic :", statistic)
print("old그룹 pvalue :", pvalue)
```

```python
#등분산성 검정
statistic, pvalue = stats.levene(new_df['충전시간'], old_df['충전시간'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

```python
statistic, pvalue = stats.ttest_ind(new_df['충전시간'], old_df['충전시간'],equal_var=True, alternative='less')
print("statistic :", statistic)
```

2.위의 통계량에 대한 p-value를 구하시오.

```python
print("pvalue :", pvalue)
```

3.유의수준 0.05하에서 귀무가설을 기준으로 검정의 결과를 채택/기각 중 선택해 입력하시오.

```python
if pvalue > 0.05:
    print("귀무가설 채택")
else:
    print("귀무가설 기각")
```

### 정규성 만족O, 등분산성 만족X 시 : stats.ttest_ind(A데이터, B데이터, equal_var=False, 대립가설 방향)

```python
statistic, pvalue = stats.ttest_ind(new_df['충전시간'], old_df['충전시간'],equal_var=False, alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### 정규성 만족X : stats.mannwhitneyu(A데이터,B데이터,대립가설 방향)

```python
statistic, pvalue = stats.mannwhitneyu(new_df['충전시간'], old_df['충전시간'], alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)
```
