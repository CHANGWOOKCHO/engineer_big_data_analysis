### SECTION08.다중 선형 회귀

한 도시의 배달 서비스 회사에서 한 달 동안의 주문 데이터를 수집했다. 배달 주문량은 할인율,온도,광고비(천원)와 같은 다양한 요인들에 영향을 받을 수 있다. 배달 서비스 회사는 이런 요인들을 고려해 주문량을 예측하고자 한다. 할인율, 온도, 광고비를 독립변수로 사용해 다중 선형 회귀 모델을 구축하고, 다음 문제에 답하시오.

```python
import pandas as pd
from statsmodels.formula.api import ols
from sklearn.metrics import mean_squared_error
```

```python
df = pd.DataFrame({
    '할인율': [28, 24, 13, 0, 27, 30, 10, 16, 6, 5, 7, 11, 11, 30, 25,
            4, 7, 24, 19, 21, 6, 10, 26, 13, 15, 6, 12, 6, 20, 2],
    '온도': [15, 34, 15, 22, 29, 30, 14, 17, 28, 29, 19, 19, 34, 10,
           29, 28, 12, 25, 32, 28, 22, 16, 30, 11, 16, 18, 16, 33, 12, 22],
    '광고비': [342, 666, 224, 764, 148, 499, 711, 596, 797, 484, 986, 347, 146, 362, 642,
            591, 846, 260, 560, 941, 469, 309, 730, 305, 892, 147, 887, 526, 525, 884],
    '주문량': [635, 958, 525, 25, 607, 872, 858, 732, 1082, 863, 904, 686, 699, 615, 893,
            830, 856, 679, 918, 951, 789, 583, 988, 631, 866, 549, 910, 946, 647, 943]
})
df.head()
```

1.할인율과 온도의 상관계수를 구하시오.(반올림하여 소수 둘째 자리까지 계산)

```python
print("상관계수 :", round(df['할인율'].corr(df['온도']),2))
```

2.모델의 결정계수를 구하시오.(반올림하여 소수 둘째 자리까지 계산)

```python
model = ols('주문량~할인율+온도+광고비', data=df).fit()
print("결정계수 :", round(model.rsquared,2))
```

3.각 변수의 회귀계수를 구하시오.(반올림하여 소수 넷째 자리까지 계산)

```python
print("회귀계수[할인율] :", round(model.params['할인율'],4))
print("회귀계수[온도] :", round(model.params['온도'],4))
print("회귀계수[광고비] :", round(model.params['광고비'],4))
```

4.모델의 절편을 구하시오.(반올림하여 소수 넷째 자리까지 계산)

```python
print("절편 :", round(model.params['Intercept'],4))
```

5.온도의 회귀 계수가 통계적으로 유의한지 검정하시오.(p-value를 반올림하여 소수 넷째자리까지 계산)

```python
print("p-value[온도] :",round(model.pvalues['온도'],4))
```

6.할인율이 10%, 온도가 20도, 광고비가 500,000원일 때 배달 주문량을 예측하시오. 

```python
new_data = pd.DataFrame({
    '할인율' : [10],
    '온도' : [20],
    '광고비' : [500],
})

pred = model.predict(new_data)
print("예측량 :", int(pred[0]))
```

7.잔차 제곱합을 구하시오.

```python
pred = model.predict(df)
df['잔차'] = df['주문량'] - pred
print("잔차 제곱합 :",sum(df['잔차']**2))
```

8.MSE를 구하시오.

```python
mse = mean_squared_error(df['주문량'], pred)
print("MSE :", mse)
```

9.각 변수 회귀계수에 대한 90% 신뢰구간을 구하시오.

```python
model.conf_int(alpha=0.1)
```

10. 새로운 데이터의 예측값의 90% 신뢰구간과 예측구간을 구하시오.
##### (할인율 15%, 온도 20도, 광고비 300,000원)

```python
new_data = pd.DataFrame({
    '할인율' : [15],
    '온도' : [25],
    '광고비' : [300],
})

pred = model.get_prediction(new_data)
result = pred.summary_frame(alpha=0.1)
result
```

11.광고비가 배달 주문량에 영향을 주는지 가설을 검정하시오.(유의 수준 0.05)

```python
pvalue = model.pvalues['광고비']
if pvalue > 0.05:
    print("귀무가설 채택(회귀계수는 0)")
else:
    print("귀무가설 기각(회귀계수 0이 아님)")
```
