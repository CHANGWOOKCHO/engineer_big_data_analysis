### SECTION01. 단일표본검정

##### 귀무가설 : 카페인 함량 >= 95
##### 대립가설 : 카페인 함량 < 95

#### 단일표본검정 순서
##### 1.정규성을 검정한다. ex) stats.shapiro(데이터)
##### 2-1. 정규성 만족 시 다음의 검정을 사용한다. ex) stats.ttest_1samp(데이터,기준값,대립가설 검정방향)
##### 2-2. 정규성을 불만족 시 다음의 검정을 사용한다. ex) stats.wilcoxon(데이터-기준값, 대립가설 검정방향)

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.DataFrame({
    'Caffeine(mg)': [
        94.2, 93.7, 95.5, 93.9, 94.0, 95.2, 94.7, 93.5, 92.8, 94.4,
        93.8, 94.6, 93.3, 95.1, 94.3, 94.9, 93.9, 94.8, 95.0, 94.2,
        93.7, 94.4, 95.1, 94.0, 93.6
    ]
})

df.head()
```

1.표본의 평균을 구하시오.

```python
print("평균 :", df['Caffeine(mg)'].mean())
```

2.Shapiro-Wilk 검정의 p-value를 구하시오.(정규성 검정)

```python
statistic,pvalue = stats.shapiro(df['Caffeine(mg)'])
print("statistic :", statistic)
print("pvalue :", pvalue)
```

3.단일 표본의 t-검정의 검정 통계량을 구하시오.

```python
statistic, pvalue = stats.ttest_1samp(df['Caffeine(mg)'],95,alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)
print("표준오차1 :", (df['Caffeine(mg)'].mean()-95)/statistic)
print("표준오차2 :", df['Caffeine(mg)'].std() / np.sqrt(len(df)))
```

4.단일 표본의 t-검정의 p-value를 구하시오.

```python
print("p-value :","{:.10f}".format(pvalue))
```

5.유의수준 0.05이하에서 귀무가설을 기준으로 검정결과를 채택/기각 중 선택해 입력하시오.

```python
if pvalue > 0.05:
    print("귀무가설 채택")
else:
    print("귀무가설 기각")
```

### 정규성 불만족 가정

```python
statistic, pvalue = stats.wilcoxon(df['Caffeine(mg)']-95, alternative='less')
print("statistic :", statistic)
print("pvalue :", pvalue)
```
