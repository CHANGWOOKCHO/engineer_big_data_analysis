### SECTION01. 상관계수

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.DataFrame({
    '키' : [150,160,170,175,165],
    '몸무게' : [42,50,70,64,56],
})

df.head()
```

#### 피어슨 상관계수계산 : corr()

```python
#전체 상관계수
df.corr()
```

```python
result = stats.pearsonr(df['몸무게'],df['키'])
print("상관계수 :", result.statistic)
print("p-value :", result.pvalue)
```

#### 스피어만 상관계수 : corr(method='spearman')

```python
df.corr(method='spearman')
```

```python
result = stats.spearmanr(df['몸무게'], df['키'])
print("상관계수 :", result.statistic)
print("p-value :", result.pvalue)
```

#### 켄달의 타우 상관계수 : corr(method='kendall')

```python
df.corr(method='kendall')
```

```python
result = stats.kendalltau(df['몸무게'], df['키'])
print("상관계수 :", result.statistic)
print("p-value :", result.pvalue)
```
