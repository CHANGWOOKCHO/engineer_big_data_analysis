```python
import pandas as pd
from statsmodels.formula.api import ols
from sklearn.metrics import mean_squared_error
```

```python
df = pd.DataFrame({
    '키': [150, 160, 170, 175, 165, 155, 172, 168, 174, 158,
          162, 173, 156, 159, 167, 163, 171, 169, 176, 161],
    '몸무게': [42, 50, 70, 64, 56, 48, 68, 60, 65, 52,
            54, 67, 49, 51, 58, 55, 69, 61, 66, 53]
})

df.head()
```

### 1.주어진 데이터로 최소제곱법을 이용한 단순 선형 회귀 모델을 구축하고 통계적 요약을 출력하시오.

```python
model = ols('키~몸무게', data=df).fit()
model.summary()

#model.rsquared       # 결정계수
#model.rsquared_adj   # 수정된 결정계수
#model.params         # 회귀계수
#model.pvalues        # 회귀계수별 p-value
#model.fvalue         # F 통계량
#model.f_pvalue       # F 검정 p-value
```

### 2.회귀모델의 결정 계수를 구하시오.

```python
print("결정계수 :", model.rsquared)
```

### 3.회귀 모델에서 회귀 계수(기울기와 절편)을 구하시오.

```python
print("기울기 :", model.params['몸무게'])
print("절편 :", model.params['Intercept'])
```

### 4.회귀 모델에서 몸무게의 회귀 계수가 통계적으로 유의한지 검정했을 때의 p-value를 구하시오.

```python
pvalue = "{:.10f}".format(model.pvalues['몸무게'])
print("p-value :", pvalue)
```

### 5.회귀 모델을 사용해 몸무게가 67일 때의 예측 키를 구하시오.

```python
new_data = pd.DataFrame({'몸무게' : [67]})
result = model.predict(new_data)
print("예측 키 :", result[0])
```

### 6.회귀 모델의 제곱합을 구하시오.

```python
df['잔차'] = df['키'] - model.predict(df['몸무게'])
print('잔차의 제곱합', sum(df['잔차']**2))
```

### 7.회귀모델의 MSE(평균제곱오차)를 구하시오.

```python
mse = (df['잔차']**2).mean()
print("MSE :", mse)
```

```python
predict = model.predict(df['몸무게'])

mse = mean_squared_error(df['키'], predict)
print("MSE :", mse)
```

### 심화1. 몸무게의 95% 신뢰구간을 구하시오.

```python
ci = model.conf_int(alpha=0.05)
ci
```

###  심화2. 몸무게가 50일 때 예측 키의 신뢰구간과 예측구간을 구하시오.

```python
new_data = pd.DataFrame({'몸무게' : [50]})
pred = model.get_prediction(new_data)
result = pred.summary_frame(alpha=0.05)

print(f'신뢰구간 : {result.loc[0,'mean_ci_lower']} ~ {result.loc[0,'mean_ci_upper']}')
print(f'예측구간 : {result.loc[0,'obs_ci_lower']} ~ {result.loc[0,'obs_ci_upper']}')
```
