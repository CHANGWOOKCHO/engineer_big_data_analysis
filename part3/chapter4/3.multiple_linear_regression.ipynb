```python
import pandas as pd
from statsmodels.formula.api import ols
from scipy import stats
from sklearn.metrics import mean_squared_error
```

```python
df = pd.DataFrame({
    '매출액': [300, 320, 250, 360, 315, 328, 310, 335, 326, 280,
            290, 300, 315, 328, 310, 335, 300, 400, 500, 600],
    '광고비': [70, 75, 30, 80, 72, 77, 70, 82, 70, 80,
            68, 90, 72, 77, 70, 82, 40, 20, 75, 80],
    '직원수': [15, 16, 14, 20, 19, 17, 16, 19, 15, 20,
            14, 5, 16, 17, 16, 14, 30, 40, 10, 50]
    })
df.head()
```

### 1.최소제곱법을 이용한 다중 선형 회귀 모델을 구축하고 통계적 요약을 출력하시오.

```python
model = ols('매출액~광고비+직원수', data=df).fit()
model.summary()
```

### 2.광고비와 매출액의 상관계수를 구하시오.

```python
print('상관계수 :', df['광고비'].corr(df['매출액']))
```

### 3.광고비와 매출액의 상관계수, p-value를 구하시오.

```python
result = stats.pearsonr(df['광고비'], df['매출액'])

print("상관계수 : ", result.statistic)
print("p-value : ", result.pvalue)
```

### 4.회귀모델의 결정 계수를 구하시오.

```python
print("결정계수 :", model.rsquared)
```

### 5.회귀 모델에서 회귀계수(기울기, 절편)을 구하시오.

```python
print("절편 :", model.params['Intercept'])
print("기울기(광고비) : ", model.params['광고비'])
print("기울기(직원수) : ", model.params['직원수'])
```

### 6.회귀 모델에서 광고비의 회귀계수가 통계적으로 유의한지 검정했을 때 p-value를 구하시오.

```python
print("p-value :", model.pvalues['광고비'])
```

### 7.광고비 50, 직원수 20인 데이터가 있을 때 구축된 회귀모델에서의 예상 매출액을 구하시오.

```python
new_data = pd.DataFrame({
    '광고비' : [50],
    '직원수' : [20],
})

pred = model.predict(new_data)
print("예상 매출액 :", pred[0])
```

### 8.회귀 모델의 잔차의 제곱합을 구하시오.

```python
df['잔차'] = df['매출액'] - model.predict(df)
sse = sum(df['잔차']**2)
print("잔차의 제곱합 :", sse)
```

### 9.회귀모델의 MSE를 구하시오.

```python
pred = model.predict(df)
mse = mean_squared_error(df['매출액'], pred)
print("mse :", mse)

mse = (df['잔차']**2).mean()
print("mse :", mse)
```

### 10.각 변수별 95%의 신뢰구간을 구하시오.

```python
conf = model.conf_int(alpha=0.05)
conf
```

### 11.광고비 45, 직원수 22일 때 95% 신뢰구간과 예측구간을 구하시오.

```python
new_data = pd.DataFrame({
    '광고비' : [45],
    '직원수' : [22],
})

pred = model.get_prediction(new_data)
summary = pred.summary_frame(alpha=0.05)

print('예측값 :', summary.loc[0, 'mean'])
print(f'신뢰구간 : {summary.loc[0,'mean_ci_lower']} ~ {summary.loc[0,'mean_ci_upper']}')
print(f'예측구간 : {summary.loc[0,'obs_ci_lower']} ~ {summary.loc[0,'obs_ci_upper']}')
```
