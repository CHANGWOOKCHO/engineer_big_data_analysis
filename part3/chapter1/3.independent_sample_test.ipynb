```python
import pandas as pd
from scipy import stats
```

#### 정규성O, 등분산성O : 독립 표본 검정
#### 정규성O, 등분산성X : 웰치의 T검정
#### 정규성X, 등분산성X : 만-휘트니 검정

```python
# 독립표본의 경우 데이터
class1 = [85, 90, 92, 88, 86, 89, 83, 87]
class2 = [80, 82, 88, 85, 84]
```

```python
#정규성 확인
print(stats.shapiro(class1))
print(stats.shapiro(class2))
```

```python
#등분산성 확인
print(stats.levene(class1,class2))
```

#### 독립 표본 양측검정
##### 귀무가설 : 반별 수학 평균 점수는 같다.
##### 귀무가설 : 반별 수학 평균 점수는 다르다.

```python
statistic, p_value = stats.ttest_ind(class1, class2, equal_var=True)
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 독립 표본 단측검정(less)
##### 귀무가설 : 반별 수학 평균 점수는 같다.
##### 귀무가설 : 2반 수학 평균 점수가 더 높다.

```python
statistic, p_value = stats.ttest_ind(class1, class2, equal_var=True, alternative='less')
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 독립 표본 단측검정(greater)
##### 귀무가설 : 반별 수학 평균 점수는 같다.
##### 귀무가설 : 1반 수학 평균 점수가 더 높다.

```python
statistic, p_value = stats.ttest_ind(class1, class2, equal_var=True, alternative='greater')
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 웰치의 T검정(equal_var=False)

```python
statistic, p_value = stats.ttest_ind(class1, class2, equal_var=False)
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 만-휘트니 검정

```python
statistic, p_value = stats.mannwhitneyu(class1, class2)
print('statistic :', statistic)
print('p_value :', p_value)
```
