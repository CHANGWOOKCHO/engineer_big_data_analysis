```python
import pandas as pd
from scipy import stats
```

```python
df = pd.DataFrame({
    'weights':[122, 121, 120, 119, 125, 115, 121, 118, 117, 127,
           123, 129, 119, 124, 114, 126, 122, 124, 121, 116,
           120, 123, 127, 118, 122, 117, 124, 125, 123, 121],
})
```

#### T-TEST 양측검정
##### 귀무가설 : 팝콘 라지 사이즈의 평균 무게는 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 평균 무게는 120g이 아니다.

```python
t_statistic, p_value = stats.ttest_1samp(df['weights'],120)
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### T-TEST 단측검정(greater)
##### 귀무가설 : 팝콘 라지 사이즈의 평균 무게는 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 평균 무게는 120g보다 크다.

```python
t_statistic, p_value = stats.ttest_1samp(df['weights'],120, alternative='greater')
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### T-TEST 단측검정(less)
##### 귀무가설 : 팝콘 라지 사이즈의 평균 무게는 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 평균 무게는 120g보다 작다.

```python
t_statistic, p_value = stats.ttest_1samp(df['weights'],120, alternative='less')
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### 샤피로-월크 검정(정규성 검정)
##### 귀무가설 : 주어진 데이터의 샘플이 정규 분포를 따른다.
##### 대립가설 : 주어진 데이터의 샘플이 정규 분포를 따르지 않는다.

```python
df = pd.DataFrame({
    'weights':[125, 126, 118, 124, 117, 127, 123, 122, 119, 142]
})
```

```python
t_statistic, p_value = stats.shapiro(df['weights'])
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### 윌-콕슨 양측검정
##### 귀무가설 : 팝콘 라지 사이즈의 중앙값이 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 중앙값은 120g이 아니다.

```python
t_statistic, p_value = stats.wilcoxon(df['weights']-120)
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### 윌-콕슨 단측검정(greater)
##### 귀무가설 : 팝콘 라지 사이즈의 중앙값이 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 중앙값은 120g보다 크다.

```python
t_statistic, p_value = stats.wilcoxon(df['weights']-120, alternative='greater')
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```

#### 윌-콕슨 단측검정(less)
##### 귀무가설 : 팝콘 라지 사이즈의 중앙값이 120g이다.
##### 대립가설 : 팝콘 라지 사이즈의 중앙값은 120g보다 작다.

```python
t_statistic, p_value = stats.wilcoxon(df['weights']-120, alternative='less')
print("t_statistic :", t_statistic)
print("p_value :", p_value)
```
