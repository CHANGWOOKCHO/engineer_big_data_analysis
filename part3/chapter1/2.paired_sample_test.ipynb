```python
import pandas as pd
from scipy import stats
```

```python
df = pd.DataFrame({
    'before':[85, 90, 92, 88, 86, 89, 83, 87],
    'after':[85.5,89.9,92.6,89.5,85.8,88.8,84.6,87.8]
})
```

#### 대응 표본 단측검정(before-after < 0)
##### 귀무가설 : 새로운 교육 프로그램은 점수 상승에 효과가 없다.
##### 대립가설 : 새로운 교육 프로그램은 점수 상승에 효과가 있다.

```python
statistic, p_value = stats.ttest_rel(df['before'], df['after'], alternative='less')
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 대응 표본 단측검정(before-after > 0)
##### 귀무가설 : 새로운 교육 프로그램은 점수 하락에 효과가 없다.
##### 대립가설 : 새로운 교육 프로그램은 점수 하락에 효과가 있다.

```python
statistic, p_value = stats.ttest_rel(df['before'], df['after'], alternative='greater')
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 대응 표본 양측검정
##### 귀무가설 : 교육 전후 차이가 없다.
##### 대립가설 : 교육 전후 차이가 있다.

```python
statistic, p_value = stats.ttest_rel(df['after'], df['before'])
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 샤피로-월크 검정(정규성 검정)
##### 귀무가설 : 주어진 데이터 샘플이 정규 분포를 따른다.
##### 대립가설 : 주어진 데이터 샘플이 정규 분포를 따르지 않는다.

```python
df = pd.DataFrame({
    'before':[85, 90, 92, 88, 86, 89, 83, 87],
    'after':[86, 92, 94, 89, 84, 90, 84, 88]
})
df['diff'] = df['after'] - df['before']   
```

```python
statistic, p_value = stats.shapiro(df['diff']) 
print('statistic :', statistic)
print('p_value :', p_value)
```

#### 윌-콕슨 부호 순위 검정

```python
statistic, p_value = stats.wilcoxon(df['after'],df['before'], alternative='greater')
print('statistic :', statistic)
print('p_value :', p_value)
```

```python
statistic, p_value = stats.wilcoxon(df['diff'], alternative='greater')
print('statistic :', statistic)
print('p_value :', p_value)
```
