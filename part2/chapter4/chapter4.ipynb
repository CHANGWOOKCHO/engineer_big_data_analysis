### 데이터 전처리

```python
#데이터 유형별 컬럼 확인
str_cols = train.select_dtypes(include='object').columns
```

```python
num_cols = train.select_dtypes(include='number').columns
```

```python
target = train.pop('Item_Outlet_Sales')
print('train :', train.shape)
print('test :', test.shape)
```

### 인코딩

```python
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.concat([train,test])
print('data :', data.shape)
```

```python
le = LabelEncoder()

for col in str_cols:
    data[col] = le.fit_transform(data[col])
```

```python
data.head()
```

### 결측치 처리

```python
from sklearn.model_selection import train_test_split
```

```python
train = data.iloc[:len(train)].copy()
test = data.iloc[len(train):].copy()

print('train 행열:', train.shape)
print('test 행열:', test.shape)
```

```python
weight = train['Item_Weight'].min()
size = train['Outlet_Size'].mode()[0]

train['Item_Weight'] = train['Item_Weight'].fillna(weight)
train['Outlet_Size'] = train['Outlet_Size'].fillna(size)

test['Item_Weight'] = test['Item_Weight'].fillna(weight)
test['Outlet_Size'] = test['Outlet_Size'].fillna(size)
```

### 검증 데이터 나누기

```python
x_train,x_val,y_train,y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0,
)
print('x_train :', x_train.shape)
print('x_val :', x_val.shape)
print('y_train :', y_train.shape)
print('y_val :', y_val.shape)
```

## 학습 및 평가

### 선형회귀

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
from sklearn.metrics import root_mean_squared_error
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)

y_pred = lr.predict(x_val)
```

```python
print('MSE :', mean_squared_error(y_val, y_pred))
print('MAE :', mean_absolute_error(y_val, y_pred))
print('R2 Score:', r2_score(y_val, y_pred))
print('RMSE :', root_mean_squared_error(y_val, y_pred))
```

### 랜덤포레스트

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(random_state=0)
rf.fit(x_train,y_train)

y_pred = rf.predict(x_val)
```

```python
print('MSE :', mean_squared_error(y_val, y_pred))
print('MAE :', mean_absolute_error(y_val, y_pred))
print('R2 Score:', r2_score(y_val, y_pred))
print('RMSE :', root_mean_squared_error(y_val, y_pred))
```

### lightGBM 

```python
import lightgbm as lgb
```

```python
model = lgb.LGBMRegressor(random_state=0, ver_bose=-1)
model.fit(x_train,y_train)

y_pred = model.predict(x_val)
```

```python
print('MSE :', mean_squared_error(y_val, y_pred))
print('MAE :', mean_absolute_error(y_val, y_pred))
print('R2 Score:', r2_score(y_val, y_pred))
print('RMSE :', root_mean_squared_error(y_val, y_pred))
```
