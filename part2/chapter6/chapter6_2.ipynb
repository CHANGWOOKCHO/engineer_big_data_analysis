```python
#기본
import numpy as np
import pandas as pd

#인코딩
from sklearn.preprocessing import LabelEncoder

#스케일링
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import MaxAbsScaler

#데이터 분할
from sklearn.model_selection import train_test_split

#예측모델
from sklearn.ensemble import RandomForestClassifier

#평가
from sklearn.metrics import f1_score
```

```python
train = pd.read_csv('creditcard_train.csv')
test = pd.read_csv('creditcard_test.csv')
target = train.pop('STATUS')
```

#### 데이터 전처리

```python
m = train['OCCUPATION_TYPE'].mode()[0]
train['OCCUPATION_TYPE'] = train['OCCUPATION_TYPE'].fillna(m)
test['OCCUPATION_TYPE'] = test['OCCUPATION_TYPE'].fillna(m)
```

```python
#컬럼(ID) 삭제
train = train.drop('ID', axis=1)
test = test.drop('ID', axis=1)
```

```python
#인코딩
obj_cols = train.select_dtypes(include='O').columns

for col in obj_cols:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    test[col] = le.transform(test[col])
```

```python
#검증 데이터 나누기
x_train,x_val, y_train, y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0,
)

print('x_train :', x_train.shape)
print('x_val :', x_val.shape)
print('y_train :', y_train.shape)
print('y_val :', y_val.shape)
```

### 학습 및 평가

```python
model = RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=0)
model.fit(x_train,y_train)
```

```python
pred = model.predict(x_val)
pred
```

```python
#평가
f1 = f1_score(y_val, pred)
print("f1 :", f1)
```

```python
pred = model.predict(test)
submit = pd.DataFrame({'pred' : pred})
submit.to_csv('credit_result.csv', index=False)
submit.head()
```
