### 성능개선

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import MaxAbsScaler
```

### 적합 스케일링 탐색

```python
scaler_dic = {
    "MinMaxScaler" : MinMaxScaler(),
    "StandardScaler" : StandardScaler(),
    "RobustScaler" : RobustScaler(),
    "MaxAbsScaler" : MaxAbsScaler(),
}
```

```python
for name,scaler in scaler_dic.items():

    train = pd.read_csv('diabetes_train.csv')
    test = pd.read_csv('diabetes_test.csv')

    target = train.pop('Outcome')

    #스케일링 모델별로 실행
    train = scaler.fit_transform(train)
    test = scaler.transform(test)

    x_train,x_val,y_train,y_val = train_test_split(
        train,
        target,
        test_size=0.2,
        random_state=0
    )
        
    model = RandomForestClassifier(random_state=0)
    model.fit(x_train,y_train)

    pred = model.predict_proba(x_val)

    roc_auc = roc_auc_score(y_val, pred[:,-1])
    
    print(name,":",roc_auc)
```

### 하이퍼파라미터 탐색

### max_depth 탐색 

```python
train = pd.read_csv('diabetes_train.csv')
test = pd.read_csv('diabetes_test.csv')

target = train.pop('Outcome')

# min-max 스케일링
scaler = MinMaxScaler()
train = scaler.fit_transform(train)
test = scaler.transform(test)

x_train,x_val,y_train,y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0
)

for depth in range(3,10) :
    model = RandomForestClassifier(max_depth=depth, random_state=0)
    model.fit(x_train,y_train) 

    pred = model.predict_proba(x_val)
    roc_auc = roc_auc_score(y_val, pred[:, -1])
    print(f'depth({depth}) : {roc_auc}')
```

### n_estimators 탐색

```python
train = pd.read_csv('diabetes_train.csv')
test = pd.read_csv('diabetes_test.csv')

target = train.pop('Outcome')

# min-max 스케일링
scaler = MinMaxScaler()
train = scaler.fit_transform(train)
test = scaler.transform(test)

x_train,x_val,y_train,y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0
)

for n in range(100,1000,100) :
    model = RandomForestClassifier(max_depth=5, n_estimators=n, random_state=0)
    model.fit(x_train,y_train) 

    pred = model.predict_proba(x_val)
    roc_auc = roc_auc_score(y_val, pred[:, -1])
    print(f'n_estimators({n}) : {roc_auc}')
```
