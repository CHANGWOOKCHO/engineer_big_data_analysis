```python
#기본
import numpy as np
import pandas as pd

#전처리(인코딩)
from sklearn.preprocessing import LabelEncoder 

#전처리(스케일링)
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import MaxAbsScaler

#전처리(검증 데이터 나누기)
from sklearn.model_selection import train_test_split

#모델학습
from sklearn.ensemble import RandomForestClassifier

#평가
from sklearn.metrics import f1_score
```

```python
train = pd.read_csv('score_train.csv')
test = pd.read_csv('score_test.csv')
```

```python
print('train :', train.shape)
print('test :', test.shape)
```

```python
#y값 분리
target = train.pop('Credit_Score')
```

### 전처리(인코딩)

```python
data = pd.concat([train, test])
data = pd.get_dummies(data)

train = data[:len(train)].copy()
test = data[len(train):].copy()

print('train :', train.shape)
print('test :', test.shape)
```

### 검증데이터 나누기(랜덤포레스트는 스케일링이 큰 의미를 가지지 않음으로 생략)

```python
x_train,x_val,y_train,y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0,
)

print('x_train :', x_train.shape)
print('x_val :', x_val.shape)
print('y_train :', y_train.shape)
print('y_val :', y_val.shape)
```

### 머신러닝 학습

```python
model = RandomForestClassifier()
model.fit(x_train,y_train)

pred = model.predict(x_val)
```

#### 머신러닝 평가

```python
f1 = f1_score(y_val, pred, average='micro')
print('f1_score :', f1)
```

```python
submit = pd.DataFrame({'pred':pred})
submit.to_csv('score_result.csv', index=False)
```
