```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
```

```python
df = pd.read_csv('design.csv')
df.head()
```

```python
cond1 = df['id'] <= 140
cond2 = df['id'] > 140

train = df[cond1]
test = df[cond2]
```

```python
model = ols('design ~ c1 + c2 + c3 + c4 + c5', data=train).fit()
model.summary()
```

```python
pvalues = model.pvalues[1:][model.pvalues < 0.05]
pvalues
```

```python
model = ols('design ~ c1 + c2 + c4', data=train).fit()
model.summary()
```

```python
train['pred'] = model.predict(train)
```

```python
corr = train['design'].corr(train['pred'])
print("피어슨 상관계수 :", round(corr,3))
```

```python
from sklearn.metrics import root_mean_squared_error
```

```python
test['pred'] = model.predict(test)
rmse = root_mean_squared_error(test['design'], test['pred'])
print("rmse :", round(rmse,3))
```

### 문제2

```python
df = pd.read_csv('retention.csv')
df.head()
```

```python
from statsmodels.formula.api import logit
```

```python
model = logit('Churn~MonthlyCharges + CustomerTenure + HasPhoneService + HasTechInsurance', data=df).fit()
model.summary()
```

```python
result = model.pvalues['MonthlyCharges']
result = round(result,3)
print("MonthlyCharges p-value :", result)
```

```python
odds = np.exp(model.params['HasPhoneService'])
print("odds :", round(odds,3))
```

```python
pred = model.predict(df)
result = sum(pred > 0.3)
print("result :", result)
```
