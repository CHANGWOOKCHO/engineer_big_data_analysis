### 문제1

```python
df = pd.read_csv('loan.csv')
df.head()
```

```python
df['총대출'] = df['신용대출']+df['담보대출']
```

```python
sum_df = df.groupby(['지역코드', '성별'])['총대출'].sum()
sum_df = sum_df.unstack()
sum_df
```

```python
sum_df['diff'] = abs(sum_df[1] - sum_df[2])
print("최대 차이의 절대값 :", sum_df['diff'].max())
print("최대 차이 지역코드 :", sum_df['diff'].idxmax())
```

### 문제2

```python
df = pd.read_csv('crime.csv')
df.head()
```

```python
df = pd.melt(df, id_vars=['연도', '구분'], var_name='범죄유형', value_name='건수')
df = df.groupby(['연도','구분','범죄유형'])['건수'].sum().unstack(level=1)
df
```

```python
df = df.reset_index(drop=False)
df
```

```python
df['검거율'] = df['검거건수']/df['발생건수']
```

```python
sum_df = df.groupby('연도')['검거율'].idxmax()
sum_df
```

```python
df.loc[sum_df]['검거건수'].sum()
```

```python
df = pd.read_csv('crime.csv')
df.head()
```

```python
df1 = df[df['구분'] == '발생건수']
df2 = df[df['구분'] == '검거건수']

df1 = df1.set_index('연도', drop=True)
df2 = df2.set_index('연도', drop=True)

df1 = df1.drop('구분', axis=1)
df2 = df2.drop('구분', axis=1)
```

```python
df3 = df2/df1
df3
```

```python
listbox = df3.idxmax(axis=1)
listbox
```

```python
sum = 0
for idx,col in listbox.items():
    cnt = df2.loc[idx,col]
    sum = sum + cnt
print(sum)
```

### 문제3

```python
df = pd.read_csv('hr.csv')
df.head()
```

```python
mean_val = df['만족도'].mean()
df['만족도'] = df['만족도'].fillna(mean_val)
```

```python
mean_df = df.groupby(['부서', '성과등급'])['근속연수'].transform('mean')
mean_df = mean_df.astype(int)
mean_df
```

```python
df['근속연수'] = df['근속연수'].fillna(mean_df)
df['연봉_근속연수'] = df['연봉']/df['근속연수']
```

```python
df.head()
```

```python
df = df.sort_values('연봉_근속연수', ascending=False)
df.head()
```

```python
var1 = df.iloc[2]['근속연수']

df['연봉_만족도'] = df['연봉']/df['만족도']
df = df.sort_values('연봉_만족도', ascending=False)
```

```python
var2 = df.iloc[1]['교육참가횟수']

result = var1+var2
```

```python
print("결과 :", int(result))
```
