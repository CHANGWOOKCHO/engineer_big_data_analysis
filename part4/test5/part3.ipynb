```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
from statsmodels.formula.api import logit
from sklearn.metrics import accuracy_score
```

### 문제1

```python
df = pd.read_csv('clam.csv')
df.head()
```

```python
train = df.iloc[:210]
test = df.iloc[210:]
```

```python
model = logit('gender~weight', data=train).fit()
model.summary()
```

1.몸무게가 증가했을 때 수컷일 오즈비값은?

```python
odds = np.exp(model.params['weight'])
odds = round(odds,4)
print("오즈비(남성) :", odds)
```

1.몸무게가 증가했을 때 여성일 오즈비값은?

```python
or_odds = 1/odds
or_odds = round(or_odds,4)
print("오즈비(여성) :", or_odds)
```

2.gender 종속변수 이외의 변수를 독립변수로 하여 모델 생성 후 모델의 잔차이탈도를 계산하시오.

```python
model = logit('gender ~ age + length + diameter + height + weight',data=train).fit()
model.summary()
```

```python
llf = model.llf
rm = round(-2 * llf,2)
print("잔차이탈도 :", rm)
```

3.독립변수 weight만 사용한 모델에서 test데이터의 gender를 예측하고 error rate(오류율)을 구하시오.(반올림하여 소수 셋째자리)

```python
model = logit('gender~weight', data=train).fit()
```

```python
pred = model.predict(test)
pred = (pred > 0.5).astype(int)

accuracy = accuracy_score(test['gender'], pred)
error_rate = 1-accuracy

print("오류율 :", round(error_rate,3))
```

### 문제2

```python
df = pd.read_csv('system_cpu.csv')
df.head()
```

1. ERP와 가장 상관관계가 높은 값을 구하시오.

```python
corr = df.corr()['ERP']
erp_corr = corr.sort_values(ascending=False).iloc[1]
print("상관계수 :", round(erp_corr,3))
```

2.cpu 100미만 데이터로 erp를 종속변수 나머지를 독립변수로 설정한 선형회귀모델을 생성하고 결정계수를 구하시오.(소수점 셋째 자리)

```python
train = df[df['CPU'] < 100]

model = ols('ERP ~ Feature1 + Feature2 + Feature3 + CPU', data= train).fit()
model.summary()
```

```python
print("결정계수 :", round(model.rsquared,3))
```

3.독립변수 중 p-value가 가장 높은 값을 구하시오.

```python
pvalues = model.pvalues.iloc[0:].sort_values(ascending=False)

print("가장 높은 pvalue :", round(pvalues.iloc[0],3))
```
