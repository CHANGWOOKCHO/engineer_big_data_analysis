```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestRegressor
from lightgbm import LGBMRegressor

from sklearn.metrics import root_mean_squared_error
```

```python
train = pd.read_csv('mart_train.csv')
test = pd.read_csv('mart_test.csv')

print("train :", train.shape)
print("test :", test.shape)
```

```python
target = train.pop('total')
```

```python
target.describe()
```

```python
train.isnull().sum()
```

```python
#columns = train.select_dtypes('O').columns

#le = LabelEncoder()
#for col in columns :
#    train[col] = le.fit_transform(train[col])
#    test[col] = le.transform(test[col])

#train.head()
cnt = len(train)
data = pd.concat([train,test])
data = pd.get_dummies(data)

train = data.iloc[:cnt].copy()
test = data.iloc[cnt:].copy()
```

```python
x_train,x_val,y_train,y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0
)
```

```python
model1 = RandomForestRegressor(random_state=0)
model1.fit(x_train,y_train)
pred = model1.predict(x_val)
```

```python
rmse = root_mean_squared_error(y_val, pred)
print("랜덤포레스트 :", rmse)
```

```python
model2 = LGBMRegressor(random_state=0, verbose=-1)
model2.fit(x_train,y_train)
pred = model2.predict(x_val)
```

```python
rmse = root_mean_squared_error(y_val, pred)
print("LGBM :", rmse)
```

```python
pred = model1.predict(test)
result = pd.DataFrame({
    'pred' : pred
})

result.to_csv('result.csv', index=False)
```
