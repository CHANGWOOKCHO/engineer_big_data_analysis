```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

문제1. 결측치가 있는 행을 제거한 후 학생이 가장 많이 수강한 과목을 찾고, 해당 과목 점수를 표준화(standardScaler)한 다음 가장 큰 표준화된 값을 구하시오.(소수점 셋째 자리)

```python
df = pd.read_csv('student_assessment.csv')
df.head()
```

```python
df = df.dropna()
```

```python
idx = df.value_counts('id_assessment').idxmax()
```

```python
cond = df['id_assessment'] == idx
df = df[cond]
```

```python
scaler = StandardScaler()
df['score_scaler'] = scaler.fit_transform(df[['score']])
df.head()
```

```python
df = df.sort_values('score_scaler', ascending=False)
result = df.iloc[0]['score_scaler']
result = round(result,3)
print("result :", result)
```

문제2. DE1~DE77 컬럼 중 주가지수의 종각 'close'와 가장 상관관계가 높은 변수를 찾아 해당 변수의 평균값을 구하시오.(소수점 넷째자리)

```python
df = pd.read_csv('stock_market.csv')
df.head()
```

```python
corr_df = df.corr()['close'].abs()
corr_df
```

```python
corr_df = corr_df.loc['DE1':'DE77']
```

```python
idx = corr_df.idxmax()
print("가장 큰 컬럼 :", idx)
```

```python
result = df[idx].mean()
result = round(result,4)
print("결과 :", result)
```

### 문제3. IQR을 이용해 이산화탄소(CO2) 이상치 수를 찾으시오.

```python
df = pd.read_csv('air_quality.csv')
df.head()
```

```python
q1 = df['CO2'].quantile(0.25)
q3 = df['CO2'].quantile(0.75)
iqr = q3-q1

print("Q1 :", q1)
print("Q3 :", q3)
print("IQR :", iqr)
```

```python
upper = q3 + (IQR*1.5)
lower = q1 - (IQR*1.5)

cond1 = df['CO2'] < lower
cond2 = df['CO2'] > upper

df = df[cond1 | cond2]
```

```python
print("이상치 수 :", len(df))
```
