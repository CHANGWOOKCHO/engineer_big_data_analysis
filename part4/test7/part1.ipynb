문제1.종량제봉투 종류가 '규격봉투'고, 종량제봉퉁용도가 '음식물쓰레기'인 2L 가격 평균을 구하시오. 단 가격이 0이면 평균에서 제외하고, 반올림 후 정수를 출력하시오.

```python
df = pd.read_csv('data5-1.csv')
df.head()
```

```python
cond1 = df['종량제봉투종류'] == '규격봉투'
cond2 = df['종량제봉투용도'] == '음식물쓰레기'
cond3 = df['2ℓ가격'] > 0
df = df[cond1 & cond2 & cond3]
```

```python
mean_val = df['2ℓ가격'].mean()
print("mean_val :", round(mean_val))
```

문제2.bmi수치가 정상인 사람 수와 위험체중인 사람 수의 차이를 절댓값으로 구하시오.
##### bmi = 몸무게/키
##### 저체중 : bmi 18.5미만
##### 정상체중 : bmi 18.5이상 ~ 23미만
##### 과체중 : bmi 23이상 ~ 25미만
##### 비만체중 : bmi 25이상

```python
df = pd.read_csv('data5-2.csv')
df.head()
```

```python
df['bmi'] = df['Weight']/(df['Height']/100) ** 2
df.head()
```

```python
cond1 = df['bmi'].between(18.5,23, inclusive='left')
cond2 = df['bmi'].between(23,25, inclusive='left')

result = abs(sum(cond1) - sum(cond2))
print("result :", result)
```

문제3.순 전입학생이 가장 많은 학교의 전체 학생수를 구하시오.(순 전입학생 = 전입학생 - 전출학생)

```python
df = pd.read_csv('data5-3.csv')
df
```

```python
df.info()
```

```python
df['순전입'] = df['전입학생수(계)'] - df['전출학생수(계)']
df = df.sort_values('순전입', ascending=False)
result = df.iloc[0]['전체학생수(계)']
print("result :", result)
```
