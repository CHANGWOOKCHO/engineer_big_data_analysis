```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier
from sklearn.metrics import f1_score
```

```python
train = pd.read_csv('game_train.csv')
test = pd.read_csv('game_test.csv')

print('train :', train.shape)
print('test :', test.shape)
```

```python
train.sample(5)
```

```python
target = train.pop('User_Level')
```

```python
target.value_counts()
```

```python
#검증데이터 분리
x_train,x_val, y_train, y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0
)

print("x_train :", x_train.shape)
print("x_val :", x_val.shape)
print("y_train :", y_train.shape)
print("y_val :", y_val.shape)
```

```python
model1 = RandomForestClassifier(random_state=0)
model1.fit(x_train,y_train)
pred1 = model1.predict(x_val)
```

```python
f1 = f1_score(y_val, pred1, average='macro')
print("랜덤포레스트 :", f1)
```

```python
model2 = LGBMClassifier(random_state=0, verbose=-1)
model2.fit(x_train,y_train)
pred2 = model2.predict(x_val)
```

```python
f1 = f1_score(y_val, pred2, average='macro')
print("lightGBM :", f1)
```

### 심화학습

```python
#검증데이터 분리
x_train,x_val, y_train, y_val = train_test_split(
    train,
    target,
    test_size=0.2,
    random_state=0,
    stratify=target
)

print("x_train :", x_train.shape)
print("x_val :", x_val.shape)
print("y_train :", y_train.shape)
print("y_val :", y_val.shape)
```

```python
model1 = RandomForestClassifier(random_state=0, class_weight='balanced')
model1.fit(x_train,y_train)
pred1 = model1.predict(x_val)
```

```python
f1 = f1_score(y_val, pred1, average='macro')
print("랜덤포레스트 :", f1)
```

```python
model2 = LGBMClassifier(random_state=0, verbose=-1,class_weight='balanced')
model2.fit(x_train,y_train)
pred2 = model2.predict(x_val)
```

```python
f1 = f1_score(y_val, pred2, average='macro')
print("랜덤포레스트 :", f1)
```

### 전체데이터로 재학습

```python
model = LGBMClassifier(random_state=0, verbose=-1,class_weight='balanced')
model.fit(train,target)
pred = model.predict(test)
```

```python
result = pd.DataFrame({
    'pred' : pred,
})
result.to_csv('result.csv', index=False)
```
