```python
import numpy as np
import pandas as pd
```

### 문제1

```python
df = pd.read_csv('greenhouse_gas.csv')
df.head()
```

```python
col = df.columns[1:]
max_index = df[col].idxmax()
country_df = df.loc[max_index, 'Country']
```

```python
rank_df = country_df.value_counts()
```

```python
print("1위를 가장 많이한 국가 :", rank_df.idxmax())
print("1위 횟수 :", rank_df.max())
```

```python
df['CAGR'] = (df['2021']/df['2001']) ** (1/20) - 1
```

```python
max_val = df['CAGR'].max()
print("CAGR 최댓값 :", round(max_val, 3))
```

### 문제2

```python
df = pd.read_csv('sensor_data.csv')
df.head()
```

```python
null_df = df.isnull().sum()

print("결측치가 가장 많은 컬럼 :", null_df.idxmax())
print("결측치 수 :", null_df.max())
```

```python
median_val = df['S2_Temp'].median()
df['S2_Temp'] = df['S2_Temp'].fillna(median_val)
```

```python
mean_val = df['S2_Temp'].mean()
print("S2_Temp 평균 :", round(mean_val,3))
```

### 문제3

```python
df = pd.read_csv('order_data.csv')
df.head()
```

```python
df['order_amt'] = df['quantity'] * df['price']
```

```python
df['cancel_TF'] = df['order_no'].str[0] == 'C'
```

```python
cond = df['cancel_TF'] == True
df.loc[cond, 'order_amt'] = df.loc[cond, 'order_amt'] * -1
df.head()
```

```python
cond = df['cancel_TF']
idx =df[cond]['order_amt'].abs().idxmax()
print("가장 큰 주문번호 :", df.loc[idx, 'order_no'])
```

```python
sum_df = df.groupby('cus_id')['order_amt'].sum()
print("가장 큰 고객 :", sum_df.idxmax())
print("가장 큰 값 :", round(sum_df.max(),2))
```
