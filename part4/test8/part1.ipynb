문제1. 'age'컬럼의 3사분위수와 1사분위수의 차이를 절댓값으로 구하고, 소수점 이하를 버리고 정수로 출력하시오.

```python
df = pd.read_csv('data4-1.csv')
df.head()
```

```python
q1 = df['age'].quantile(0.25)
q3 = df['age'].quantile(0.75)

iqr = abs(q3-q1)
print("IQR :", int(iqr))
```

문제2. 모든 reactions 중 loves와 wows를 합한 반응 비율이 40%보다 크고 50%보다 작은 데이터를 찾고, 이 중 type컬럼이 'video'인 데이터의 개수를 구하시오.

```python
df = pd.read_csv('data4-2.csv')
df.head()
```

```python
df['radio'] = (df['loves'] + df['wows']) / df['reactions'] 
```

```python
cond1 = df['radio'].between(0.4, 0.5, inclusive='neither')
cond2 = df['type'] == 'video'
df =df[cond1 & cond2]

print("result :", len(df))
```

문제3. 'date_added'컬럼의 날자가 2018년1월이면서 'country'에서 United Kingdom 단독 제작인 데이터의 개수를 구하시오.

```python
df = pd.read_csv('data4-3.csv')
df.head()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
df['date'] = df['date_added'].dt.to_period('M')
df.head()
```

```python
cond1 = df['date'] == '2018-01'
cond2 = df['country'] == 'United Kingdom'

df = df[cond1 & cond2]
print("result :", len(df))
```
