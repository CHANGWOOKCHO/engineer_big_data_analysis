```python
import numpy as np
import pandas as pd

from scipy import stats
```

1.감기약 위약을 투여받은 환자의 부작용은 항암약 위약을 투여받은 환자의 부작용과 차이가 있나? 

```python
df = pd.DataFrame({
    "항암약":[4,4,3,4,1,4,1,4,1,4,4,2,1,4,2,3,2,4,4,4]
    })
print(df.head(3))
```

1. 4(이상없음)의 비율을 0과 1사이의 값으로 구하시오.

```python
radio_df = df['항암약'].value_counts(normalize=True)
print(radio_df.loc[4])
```

2. 감기약의 예상 부작용 비율과 항암약의 부작용 관찰값이 통계적으로 유의미한 차이가 있는지 확인하기 위해 카이제곱을 사용해 검정통계량을 구하시오.

```python
prob_list = [0.1,0.05,0.15,0.7]
cnt = len(df)

for i in range(len(prob_list)):
    prob_list[i] = prob_list[i] * cnt

#기대빈도
prob_list
```

```python
obs_list = df['항암약'].value_counts().sort_index().to_list()
obs_list
```

```python
statistic, pvalue = stats.chisquare(obs_list, prob_list)
print("statistic :", statistic)
print("pvalue :", pvalue)
```

### 문제3 

```python
df = pd.read_csv('data6-3-2.csv')
df.head()
```

```python
from statsmodels.formula.api import ols
```

```python
model = ols('temperature~solar + wind + o3', data=df).fit()
print(model.summary())
```

```python
print('o3 회귀계수 :', model.params['o3'])
```

```python
print("wind pvalue :", model.pvalues['wind'])
```

```python
new_data = pd.DataFrame({
    'solar' : [100],
    'wind' : [5],
    'o3' : [30],
})
```

```python
pred = model.predict(new_data)
print("예측값 :", pred[0])
```
