```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import logit
from statsmodels.formula.api import ols
```

### 문제1

```python
df = pd.read_csv('churn.csv')
df.head()
```

##### 1.로지스틱 회귀 분석을 수행해 유의활률 0.05이상의 유의하지 않은 독립변수의 개수를 구하시오.

```python
cond = 'Churn ~ AccountWeeks + ContractRenewal + DataPlan + DataUsage + CustServCalls + DayMins + DayCalls + MonthlyCharge + OverageFee + RoamMins'
model = logit(cond, data=df).fit()

model.summary()
```

```python
print('유의하지 않은 변수 :', sum(model.pvalues >= 0.05))
```

##### 2. 유의확률 0.05미만인 유의한 변수만을 사용해 다시 로지스틱 회귀분석을 수행하시오. 이 회귀식의 유의한 회귀계수의 합계를 구하시오. (반올림 소수 셋째자리)

```python
model.pvalues[model.pvalues < 0.05]
```

```python
model = logit('Churn ~ DataUsage + DayMins', data=df).fit()
model.summary()
```

```python
sum_val = model.params.sum()
print("합계 :", round(sum_val,3))
```

##### 3.'DataUsage' 변수가 5만큼 증가할 때 오즈비를 구하시오.(반올림 소수 셋째자리)

```python
result = np.exp(model.params['DataUsage']*5)
result = round(result, 3)
print("result : ", result)
```

### 문제2

```python
df = pd.read_csv('piq.csv')
df.head()
```

1.종속변수(PIQ)로 다중 선형회귀 분석을 수행하시오. 이 때 유의확률이 가장 작은 변수의 회귀 계수값을 구하시오.(소수점 셋째 자리)

```python
model = ols('PIQ ~ Brain+Height+Weight',data=df).fit()
model.summary()
```

```python
idx = model.pvalues[1:].idxmin()
result = model.params.loc[idx]
result = round(result,3)
print("유의수준이 가장 작은 컬럼의 회귀계수 :", result)
```

2.모델의 결정계수를 구하시오.(소수점 둘째 자리)

```python
print("모델 결정계수 : ", round(model.rsquared,2))
```

##### 3.뇌크기 90, 키 70, 몸무게150 일 때 PIQ를 예측하시오.(반올림하여 정수)

```python
new_data = pd.DataFrame({
    'Brain': [90],
    'Height' : [70],
    'Weight' : [150],
})
```

```python
pred = model.predict(new_data)
result = round(pred[0])
print("예측값 :", result)
```
