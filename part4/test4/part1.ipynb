### 문제1

```python
df = pd.read_csv('drinks.csv')
df.head()
```

##### 1.대륙별(continent) 맥주 소비량(beer servings) 평균을 계산하고 평균이 가장 큰 대륙을 찾으시오.

```python
group_df = df.groupby('continent')['beer_servings'].mean()
max_continent = group_df.idxmax()
print("가장 큰 대륙 :", max_continent)
```

##### 2.찾은 대륙에서 맥주 소비량이 5번째로 많은 국가의 맥주소비량을 구하시오.

```python
cond = df['continent'] == 'Europe'
df = df[cond]

df = df.sort_values('beer_servings', ascending=False)
```

```python
print('맥주소비량 :', df.iloc[4]['beer_servings'])
```

### 문제2

```python
df = pd.read_csv('tourist.csv')
df.head()
```

##### 1. '관광객비율'이 두번재로 높은 국가의 '사업' 방문객 수를 a라고 정의하시오.
##### 관광객 비율 = 관광/(관광 + 공무 + 사업 + 기타)

```python
df['관광객비율'] = df['관광'] / (df['관광'] + df['공무'] + df['사업'] + df['기타'])

df = df.sort_values('관광객비율', ascending=False)
a = df.iloc[1]['사업']
print("a :", a)
```

##### 2.'관광'이 두번째로 높은 국가의 '공무' 방문객 수를 b라고 정의하시오.

```python
df = df.sort_values('관광', ascending=False)
b = df.iloc[1]['공무']
print("b :", b)
```

##### 3.a+b의 합을 구하시오.

```python
print("result :", a+b)
```

### 문제3

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('chem.csv')
df.head()
```

##### 1.주어진 데이터에서 'co', 'nmhc' 컬럼을 각각 min-max 스케일링하시오.

```python
scaler = MinMaxScaler()

df[['co', 'nmhc']] = scaler.fit_transform(df[['co', 'nmhc']])

df.head()
```

##### 2.스케일링된 'co', 'nmhc' 컬럼의 표준편차를 구하시오.

```python
co_std = df['co'].std()
nmhc_std = df['nmhc'].std()

print("co :", co_std)
print("nmhc :", nmhc_std)
```

##### 3.'co'컬럼의 표준편차에서 'nmhc'컬럼의 표준편차를 뺀 값을 소숫점 3자리로 반올림하여 구하시오.

```python
diff = round(co_std - nmhc_std,3)
print("차이값 :", diff)
```
