### 문제1

```python
df = pd.read_csv('subject_performance.csv')
df.head()
```

```python
topic_df = df.groupby(['sub_topic'])['is_correct'].mean()
topic_df
```

```python
topic_df = topic_df.sort_values(ascending=False).drop_duplicates()
print("3번째 값 :", round(topic_df.iloc[2],3))
```

### 문제2

```python
df = pd.read_csv('cafe_sales.csv')
df.head()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='%Y.%m.%d %H:%M')
```

```python
df['period'] = df['order_date'].dt.to_period('M')
```

```python
sum_df = df.groupby(['period'])['price'].sum()
sum_df = sum_df.sort_values(ascending=False)
sum_df
```

```python
print("2번째 매출액 :", sum_df.iloc[1])
```

```python
print("4번째 매출액 :", sum_df.index[3])
```

```python
cond = df['period'] == '2024-10'
category_df = df[cond].groupby(['category'])['price'].sum()
category_df
```

```python
print("카테고리 :", category_df.idxmax())
print("매출액 :", category_df.max())
```

### 문제3

```python
df = pd.read_csv('hamspam.csv')
df.head()
```

```python
#단어 개수
df['cnt'] = df['text'].str.split(' ').str.len()
df['cnt'].value_counts()
```

```python
mean_df = df.groupby(['label'])['cnt'].mean()

diff = abs(mean_df.iloc[0] - mean_df.iloc[1])
diff = round(diff, 3)
print("차이값 :", diff)
```
