```python
import pandas as pd
import numpy as np
from statsmodels.formula.api import logit
from statsmodels.formula.api import ols
```

### 문제1

```python
df = pd.read_csv('attrition.csv')
df.head()
```

```python
model = logit('attrition ~ age + income + overtime', data=df).fit()
model.summary()
```

```python
cond = model.pvalues < 0.05
pvalue = model.pvalues[cond][1:]
```

```python
statistic = round(model.params[pvalue.index], 3)
print("회귀계수 :", statistic)
```

```python
odds = np.exp(model.params['age'])
odds = round(odds, 3)
print("오즈비 :", odds)
```

```python
new_data = pd.DataFrame({
    'age' : [40],
    'income' : [4500],
    'overtime' : [1],
})
```

```python
pred = model.predict(new_data)
result = round(pred[0],3)
print("예측결과 :", result)
```

### 문제2

```python
df = pd.read_csv('heating.csv')
df.head()
```

```python
model = ols('heating_load ~ wall + roof + glazing + height', data=df).fit()
model.summary()
```

```python
statistic = model.params[1:]
print("회귀계수의 합 :", round(statistic.sum(),3))
```

```python
cond = model.pvalues < 0.05
pvalue = model.pvalues[cond][1:]
pvalue
```

```python
model = ols('heating_load ~ roof + glazing + height', data=df).fit()
model.summary()
```

```python
print("결정계수 :", round(model.rsquared,3))
```

```python
new_data = pd.DataFrame({
    'wall' : [20],
    'roof' : [150],
    'glazing' : [20],
    'height' : [5],
})
```

```python
pred = model.predict(new_data)

result = round(pred[0],3)
print("예측값 :", result)
```
